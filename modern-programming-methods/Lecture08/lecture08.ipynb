# Introduction to Pandas
[Pandas](https://pandas.pydata.org/) is an open source library built on top of `NumPy`.
 The name is derived from the term "panel data", an econometrics term for data sets that include observations over multiple time periods for the same individuals.

- It supports fast data analysis, cleaning, and preparation, making it a key tool for preprocessing data in machine learning pipelines.
- It offers high performance and boosts productivity for users.
- It includes built-in visualisation capabilities.
- It can handle data from a wide variety of sources.
- It allows importing data in multiple formats, including CSV, Excel, and HTML.
- It provides a range of data manipulation operations, such as groupby, join, merge, melt, and concatenation, along with data cleaning functions like filling, replacing, or imputing missing values.
- It is widely used for time series analysis.

##  Series
</div>

Pandas Series are built on top of the NumPy array object, providing additional functionality.
- A Series can be indexed by both integer position and label, unlike NumPy arrays, which can only be indexed by integer position. This allows for more intuitive and flexible data access, especially when working with labelled data.
- Like a NumPy array, a Pandas Series can store any type of data, not just numeric values: strings, dates, or even arbitrary Python objects. What pandas adds are the tools for working with them, from string methods to dates and missing values.

Let’s explore this concept with some practical examples to better understand how Series work in pandas:

### Creating a Series in Different Ways

You can create a Pandas `Series` object from various data structures, including a `list`, `NumPy array`, or `dictionary`. This flexibility allows you to easily convert different types of data into a Series for further analysis.

````python
# Objects we'll convert into pd.Series
labels = ['x', 'y', 'z']
my_list = [100, 200, 300] # python list
arr = np.array([100, 200, 300]) # numpy array
dic = {'x':100, 'y':200, 'z':300}  # python dictionary
````

**Creation from List**

````python
pd.Series(data=my_list) 
````

At first glance, a Pandas Series looks a lot like a NumPy array. However, the key difference is that a Series comes with an associated index. For example, instead of just having default integer positions like 0, 1, 2 alongside the data (e.g., 100, 200, 300), you can explicitly define the index labels to be anything you want. This flexibility allows for more meaningful indexing and easier access to data.

````python
pd.Series(data=my_list, index=labels)
````

**Creation from NumPy array**

````python
pd.Series(arr)
````

````python
pd.Series(arr, labels)
````

**Creation from Dictionary**

````python
pd.Series(dic)
````

### Index-based Selection

Pandas indexing follows two paradigms. The first is **index-based selection**, which involves selecting data based on its numerical position. The `iloc` operator is used for this.

The second paradigm is **label-based selection**, used by the `loc` operator. In this case, selection is based on the data's index labels, not their positions.

Both `loc` and `iloc` are **row-first, column-second**, like indexing a 2-D NumPy array (`a[i, j]`, Lecture 2). This is the opposite of the indexing operator `[]` on a DataFrame, which selects a column first: `df['Y']['A']` is column `Y`, row `A`.

````python
df.loc['A']
````

````python
df.loc['B', 'Y']
````

````python
df.loc[['A', 'B'], ['Z', 'Y']]
````

````python
df.iloc[0, :]  # or df.iloc[0]
````

````python
df.iloc[:, 0]
````

````python
df.iloc[:3, 0]
````

````python
df.iloc[[1, 2], 0]
````

#### A "gotcha": `loc` slices include the end

`iloc` slices the way Python and NumPy do: the end of the range is left out, so `df.iloc[0:2]` gives the rows at positions 0 and 1. `loc` slices by label, and **includes** the end: `df.loc['A':'C']` gives rows `A`, `B` **and** `C`. This is easy to trip over when the index is the default 0, 1, 2, ...: `df.iloc[0:100]` is 100 rows, but `df.loc[0:100]` is 101.

````python
df.iloc[0:2]
````

````python
df.loc['A':'C']
````

### Dropping a Column

If you need to remove a column from a DataFrame, you can use the `drop()` method. This method allows you to specify the column name and set `axis=1` to indicate that you are dropping a column (since `axis=0` would drop rows).

For example, the following code will remove the column named `'new'` from the DataFrame:

````python
df.drop('new', axis=1)
````

````python
df  # note the original DataFrame is unchanged
````

#### Dropping a Column In-Place

By default, the `drop()` method returns a new DataFrame with the specified column removed, but it doesn't modify the original DataFrame. To modify the DataFrame directly, you can use the `inplace=True` argument.

With `inplace=True`, `drop()` returns `None`, so don't also assign its result: `df = df.drop('new', axis=1, inplace=True)` would leave `df` as `None`.

For example, the following code will remove the column `'new'` from the original DataFrame without creating a copy:

````python
df.drop('new', axis=1, inplace=True)
df
````

As mentioned above, rows can be dropped by passing `axis=0`:

### Conditional Selection

One of the powerful features of Pandas is **conditional selection**, which allows you to filter data based on conditions. This is very similar to how conditional selection works in NumPy (boolean masks, Lecture 2).

````python
df>0
````

````python
df[df>0]
````

You can use bracket notation to select rows where a certain condition is met. For example, if you have a DataFrame `df` and you want to select rows where a column `col` is greater than a certain value, you can do this:

````python
df[df['Z']>0]
````

````python
df[df['Z']>0]['Y']
````

````python
df[df['Z']>0][['Y','X']]
````

#### Selecting is fine; assigning needs a single `.loc`

Chaining selections like this is fine for *reading* data. Don't chain them to *change* data, though. In pandas 3, the result of every selection (a column, some rows, ...) behaves as a **copy** of the original DataFrame: this is called *Copy-on-Write*. So in a chained assignment, the last `[]` changes a temporary copy, and the original DataFrame is left as it was. Pandas warns you when you do this, with a `ChainedAssignmentError`:

````python
marks = pd.DataFrame({'Person': ['Bob', 'Sam', 'Amy'], 'Marks': [70, 45, 80]})
marks[marks['Marks'] < 50]['Marks'] = 50   # chained assignment: changes a temporary copy, not marks
marks
````

To change the original, select the rows and the column in **one** step, with `.loc[rows, columns]`:

````python
marks.loc[marks['Marks'] < 50, 'Marks'] = 50   # one step: changes marks itself
marks
````

The same goes for a selection you keep in a variable: after `low = marks[marks['Marks'] < 60]`, changing `low` never changes `marks`.

Before pandas 3, whether a selection was a copy or a *view* of the original (as NumPy slices are, Lecture 2) depended on the details, and pandas gave a `SettingWithCopyWarning` when it couldn't tell. You will still see that warning, and `.copy()` calls added to silence it, in older code and answers online. Neither is needed in pandas 3.

#### Multiple Conditions

To filter data based on two (or more) conditions, you can use the `|` (for "or") and `&` (for "and") operators. It's important to enclose each condition in parentheses to ensure proper evaluation due to operator precedence.

When using `|` and `&`, make sure to wrap each condition in parentheses to avoid errors and ensure the correct logic is applied.

````python
df[(df['X'] > 0) & (df['Y'] > 0)]
````

````python
df[(df['X'] > 0) & (df['Y'] > -0.5)]
````

````python
df
````

### Built-in Conditional Selectors

Pandas provides several built-in methods for conditional selection, making data filtering more intuitive. Two common selectors are `isin` and `isnull` (with `notnull` as its complement).

The first is **`isin`**. This method allows you to select data whose values are part of a specified list or collection. It’s particularly useful when filtering based on multiple specific values.

````python
# simple example from the Pandas documentation
df2 = pd.DataFrame({'num_legs': [2, 4], 'num_wings': [2, 0]},
                   index=['falcon', 'dog'])
df2
````

````python
df2.isin([0, 2])
````

The second is **`isnull`** (and its companion **`notnull`**). These methods are used to highlight or filter values that are missing (`NaN`) or that exist. For instance, if you want to filter out rows where certain data is missing (e.g., wines with no price in a dataset), you would use these methods.

Using these conditional selectors helps streamline data manipulation tasks and ensures cleaner, more concise filtering.

For example, to filter out rows where column `Z` has missing values, you can use the `notnull()` method.

````python
# Let's first add a row with a null (NaN) value,
# to show how we can filter the row out
df.loc['E'] = [1.0, 2.0, np.nan]
df
````

````python
print(f"isnull:\n{df['Z'].isnull()}\n")
print(f"notnull:\n{df['Z'].notnull()}")
````

````python
# Keep only the rows where Z is not null. 
# Note in pandas 2, a selection behaves like a view, not a copy, so a .copy() was needed.
# In pandas 3, no .copy() is needed: a selection behaves as a copy.
df = df.loc[df['Z'].notnull()]
df
````

### Maps

A **map** is a term borrowed from mathematics, referring to a function that transforms one set of values into another. In data science, we often need to create new representations from existing data or convert data from its current format into a desired format. Maps facilitate this process, making them crucial for efficiently handling data transformations.

There are two mapping methods you will frequently encounter.

**`map()`** is the first and simpler method, used for element-wise operations on a Series. For example, if we wanted to adjust scores so that their mean becomes 0, we can achieve this with the following approach:

````python
df_mean = df['col1'].mean()
df['col1'].map(lambda p: p - df_mean)
````

## Groupby
</div>

The `groupby()` method allows you to group rows of data together and apply aggregate functions:

````python
# Create dataframe
df = pd.DataFrame({'Course': ['NPP', 'NPP', 'EDMS', 'EDMS', 'CM', 'CM'],
                   'Person': ['Bob', 'Sam', 'Amy', 'Vanessa', 'Carl', 'Sarah'],
                   'Marks': [70, 75, 80, 65, 60, 90]},
                  index=['A', 'B', 'C', 'D', 'E', 'F'])
df
````

Now you can use the **`.groupby()`** method to group rows together based on a column name. For instance, let's group based on **Course**. This will create a **DataFrameGroupBy** object.

````python
by_course = df.groupby('Course')
print(by_course)
````

Once you have grouped your data using the **`.groupby()`** method, you can call aggregate functions on the resulting object to perform calculations on each group.

For example, after grouping by **Course**, you can calculate the mean of the values for each group by calling an aggregate method like `.mean()`:

````python
by_course.mean(numeric_only=True)
````

Another **`.groupby()`** method worth mentioning is **`agg()`**, which allows you to apply multiple aggregation functions to your DataFrame simultaneously. This is particularly useful when you want to generate a variety of summary statistics in one step.

For example, you can generate a simple statistical summary of the dataset as follows:

````python
df.groupby(['Course']).agg(['count', 'min', 'max'])
````

## Merging, Joining, and Concatenating
</div>


There are 3 main ways of combining DataFrames together: Merging, Joining and Concatenating. In this lecture we will discuss these 3 methods with examples.



````python
df1 = pd.DataFrame({'A': ['A0', 'A1', 'A2', 'A3'],
                    'B': ['B0', 'B1', 'B2', 'B3'],
                    'C': ['C0', 'C1', 'C2', 'C3'],
                    'D': ['D0', 'D1', 'D2', 'D3']},
                    index=[0, 1, 2, 3])
````

````python
df2 = pd.DataFrame({'A': ['A4', 'A5', 'A6', 'A7'],
                    'B': ['B4', 'B5', 'B6', 'B7'],
                    'C': ['C4', 'C5', 'C6', 'C7'],
                    'D': ['D4', 'D5', 'D6', 'D7']},
                    index=[4, 5, 6, 7]) 
````

````python
df3 = pd.DataFrame({'A': ['A8', 'A9', 'A10', 'A11'],
                    'B': ['B8', 'B9', 'B10', 'B11'],
                    'C': ['C8', 'C9', 'C10', 'C11'],
                    'D': ['D8', 'D9', 'D10', 'D11']},
                    index=[8, 9, 10, 11])
````

### Concatenation

Concatenation basically glues together DataFrames. Along the other axis, pandas lines the DataFrames up by their labels and fills any gaps with `NaN` (as in the second example below, which glues them side by side although their row labels don't match). You can use **pd.concat** and pass in a list of DataFrames to concatenate together:

````python
pd.concat([df1, df2, df3])
````

````python
pd.concat([df1, df2, df3],axis=1)
````

### Merging

The **merge** function allows you to merge DataFrames together using a similar logic as merging SQL Tables together. For example:

````python
left = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3'],
                       'A': ['A0', 'A1', 'A2', 'A3'],
                       'B': ['B0', 'B1', 'B2', 'B3']})
   
right = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3'],
                        'C': ['C0', 'C1', 'C2', 'C3'],
                        'D': ['D0', 'D1', 'D2', 'D3']})    
````

````python
pd.merge(left, right, how='inner', on='key')
````

````python
pd.merge?
````

````python
left = pd.DataFrame({'key1': ['K0', 'K0', 'K1', 'K2'],
                     'key2': ['K0', 'K1', 'K0', 'K1'],
                        'A': ['A0', 'A1', 'A2', 'A3'],
                        'B': ['B0', 'B1', 'B2', 'B3']})
    
right = pd.DataFrame({'key1': ['K0', 'K1', 'K1', 'K2'],
                      'key2': ['K0', 'K0', 'K0', 'K0'],
                         'C': ['C0', 'C1', 'C2', 'C3'],
                         'D': ['D0', 'D1', 'D2', 'D3']})
````

````python
# Intersection (inner) key match
pd.merge(left, right, on=['key1', 'key2'])
````

````python
# Union (outer) key match
pd.merge(left, right, how='outer', on=['key1', 'key2'])
````

````python
# All keys from right DataFrame
pd.merge(left, right, how='right', on=['key1', 'key2'])
````

````python
# All keys from left DataFrame
pd.merge(left, right, how='left', on=['key1', 'key2'])
````

### Joining
Joining is a convenient method for combining the columns of two potentially differently-indexed DataFrames into a single result DataFrame.

````python
left = pd.DataFrame({'A': ['A0', 'A1', 'A2'],
                     'B': ['B0', 'B1', 'B2']},
                     index=['K0', 'K1', 'K2']) 

right = pd.DataFrame({'C': ['C0', 'C2', 'C3'],
                      'D': ['D0', 'D2', 'D3']},
                      index=['K0', 'K2', 'K3'])
````

````python
left.join(right)
````

````python
left.join(right, how='outer')
````

### Parsing dates

### Sampling

We can randomly sample the dataframe using `df.sample`:

````python
rand_df = df.sample(n=5, random_state=rng)  # our generator: reproducible
rand_df
````

### Resample to a different time frequency

We can resample our DataFrame using the `.resample` method, which is a DateTime-based Groupby function. Try the cell below, then change `rule="D"` to `6h`, `W`, `ME`, `YE` to see how it changes.

````python
df.resample(rule="D").mean()  # daily mean
# like grouping by date, but can fill in gaps
````

We can also use other statistical functions, e.g.:

````python
dfm = df["Hs"].resample("ME")
dfm.max().head(10)  # monthly max of a Series
````

````python
df.asfreq("ME").head()  # just the temporal values with no aggregation
````
