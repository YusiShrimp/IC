### Q1D: Adding points while preserving the quadratic interpolant

The quadratic interpolant from Q1C is P2(x) = -(5/7)x^2 + (17/14)x + 1/2. Choose any additional, distinct x-coordinates and calculate each new y-value from this same formula. The original polynomial then passes through every added point and remains an interpolant of the expanded dataset.

The examples below add three points and nine points. The plots show the original points and the added points on the same quadratic curve. If an added point has a y-value different from P2(x), the Q1C polynomial will not interpolate that point.

```python
import numpy as np
import matplotlib.pyplot as plt

def p2(x):
    return -(5.0 / 7.0) * x**2 + (17.0 / 14.0) * x + 0.5

x_base = np.array([0.0, 1.0, 0.3])
y_base = p2(x_base)

x_extra_1 = np.array([0.1, 0.5, 0.8])
y_extra_1 = p2(x_extra_1)
x_extra_2 = np.linspace(0.05, 0.95, 9)
y_extra_2 = p2(x_extra_2)

x_curve = np.linspace(0.0, 1.0, 400)
fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharey=True)

examples = [
    (axes[0], x_extra_1, y_extra_1, "Three additional points"),
    (axes[1], x_extra_2, y_extra_2, "Nine additional points"),
]

for ax, x_extra, y_extra, title in examples:
    ax.plot(x_curve, p2(x_curve), label=r"$P_2(x)$ from Q1C")
    ax.scatter(x_base, y_base, color="black", zorder=3, label="Original points")
    ax.scatter(x_extra, y_extra, marker="^", color="tab:orange", zorder=3, label="Added points")
    ax.set_xlabel("$x$")
    ax.set_title(title)
    ax.grid(True, alpha=0.3)
    ax.legend()

axes[0].set_ylabel("$y$")
fig.suptitle("Additional points generated from the Q1C quadratic")
fig.tight_layout()
plt.show()

x_all_1 = np.concatenate((x_base, x_extra_1))
y_all_1 = np.concatenate((y_base, y_extra_1))
x_all_2 = np.concatenate((x_base, x_extra_2))
y_all_2 = np.concatenate((y_base, y_extra_2))
print(f"All 6 points in the first dataset lie on P_2: {np.allclose(y_all_1, p2(x_all_1))}")
print(f"All 12 points in the second dataset lie on P_2: {np.allclose(y_all_2, p2(x_all_2))}")
```

### Q1E: Least-squares quadratic fit

When additional data points do not lie on the Q1C interpolant, that polynomial cannot interpolate all the data. Here I retain the original three points and add two observations that deviate from the Q1C quadratic. I fit a quadratic by choosing coefficients that minimise the sum of squared residuals.

For each observation, the model is a*x^2 + b*x + c. This gives a linear least-squares problem A*beta approximately equals y, where each row of A is [x^2, x, 1] and beta = [a, b, c]. I use NumPy least squares to compute the coefficients. The plot compares the Q1C interpolant with the new fit, and the residual plot shows the differences between the observations and the fitted values.

```python
import numpy as np
import matplotlib.pyplot as plt

x_data = np.array([0.0, 0.3, 0.5, 0.7, 1.0])
y_data = np.array([0.5, 0.8, 1.02, 0.90, 1.0])

A = np.column_stack((x_data**2, x_data, np.ones_like(x_data)))
coefficients, residual_sum, rank, singular_values = np.linalg.lstsq(A, y_data, rcond=None)
a, b, c = coefficients
y_fit = A @ coefficients
residuals = y_data - y_fit
sse = np.sum(residuals**2)

print(f"Least-squares quadratic: y = {a:.6f}x^2 + {b:.6f}x + {c:.6f}")
print(f"Sum of squared residuals = {sse:.6f}")
print(f"Maximum absolute residual = {np.max(np.abs(residuals)):.6f}")

x_curve = np.linspace(0.0, 1.0, 400)
y_q1c = -(5.0 / 7.0) * x_curve**2 + (17.0 / 14.0) * x_curve + 0.5
y_least_squares = a * x_curve**2 + b * x_curve + c

fig, axes = plt.subplots(2, 1, figsize=(8, 8), sharex=True, gridspec_kw={"height_ratios": [3, 1]})
axes[0].scatter(x_data, y_data, color="black", zorder=3, label="Observed data")
axes[0].plot(x_curve, y_q1c, "--", label="Q1C interpolating quadratic")
axes[0].plot(x_curve, y_least_squares, label="Least-squares quadratic")
axes[0].set_ylabel("y")
axes[0].set_title("Quadratic least-squares fit to non-interpolating data")
axes[0].grid(True, alpha=0.3)
axes[0].legend()
axes[1].axhline(0.0, color="black", linewidth=1)
axes[1].scatter(x_data, residuals, color="tab:red", zorder=3)
axes[1].set_xlabel("x")
axes[1].set_ylabel("Residual")
axes[1].grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

### Q1F: Comparing quadratic fits using MSE and MAE

I construct two datasets, each containing five points and neither exactly described by a quadratic. The first has small deviations from a quadratic trend, so its fitted curve should look visually good. The second has one large outlier, which can pull the least-squares fit away from the other observations.

For each dataset I fit a quadratic by least squares and calculate the mean squared error (MSE) and mean absolute error (MAE). MSE is the mean of the squared residuals, while MAE is the mean of their absolute values. Squaring gives large residuals much greater influence, so an outlier can increase MSE disproportionately. In this example, the outlier dataset has about 716 times the MSE but about 29 times the MAE of the visually good dataset. This shows how MAE presents a less disproportionate increase than MSE, although both metrics indicate that the outlier fit is worse.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([0.0, 1.0, 2.0, 3.0, 4.0])
quadratic_trend = 1.0 + 0.5 * x + 0.2 * x**2
y_good = quadratic_trend + np.array([0.05, -0.04, 0.03, -0.02, 0.04])
y_outlier = quadratic_trend + np.array([0.05, -0.04, 0.03, -0.02, 5.0])

datasets = {"Visually good fit": y_good, "Fit with one outlier": y_outlier}
x_curve = np.linspace(x.min(), x.max(), 400)
results = {}
fig, axes = plt.subplots(2, 2, figsize=(12, 9))

for column, (name, y) in enumerate(datasets.items()):
    A = np.column_stack((x**2, x, np.ones_like(x)))
    coefficients, _, _, _ = np.linalg.lstsq(A, y, rcond=None)
    y_fit = A @ coefficients
    residuals = y - y_fit
    mse = np.mean(residuals**2)
    mae = np.mean(np.abs(residuals))
    results[name] = (mse, mae)
    a, b, c = coefficients

    axes[0, column].scatter(x, y, color="black", zorder=3, label="Data")
    axes[0, column].plot(x_curve, a * x_curve**2 + b * x_curve + c, label="Least-squares quadratic")
    axes[0, column].set_title(name)
    axes[0, column].set_xlabel("x")
    axes[0, column].set_ylabel("y")
    axes[0, column].grid(True, alpha=0.3)
    axes[0, column].legend()

    axes[1, column].axhline(0.0, color="black", linewidth=1)
    axes[1, column].scatter(x, residuals, color="tab:red", zorder=3)
    axes[1, column].set_xlabel("x")
    axes[1, column].set_ylabel("Residual")
    axes[1, column].grid(True, alpha=0.3)

axes[0, 1].annotate("outlier", xy=(x[-1], y_outlier[-1]), xytext=(-45, -25), textcoords="offset points", arrowprops={"arrowstyle": "->"})
fig.suptitle("Quadratic least-squares fits and residuals")
fig.tight_layout()
plt.show()

for name, (mse, mae) in results.items():
    print(f"{name}: MSE = {mse:.6f}, MAE = {mae:.6f}")
```

### Q1G: Overfitting and extrapolation

To illustrate overfitting, I generate eight observations from a known smooth function, f(x) = sin(pi*x), and add small deterministic perturbations. A degree-seven polynomial can interpolate all eight observations exactly, including their perturbations. A degree-three least-squares polynomial cannot pass through every point, but may capture the underlying trend more smoothly.

The left plot compares the fits within the data interval. The right plot extends the horizontal range beyond the observed data to illustrate extrapolation. In a real application the underlying function is not known, so a model that agrees with observed points cannot by itself guarantee accurate predictions outside the data range. The reported errors against f are possible here only because this is a synthetic example with a known generating function.

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.sin(np.pi * x)

x_data = np.linspace(-1.0, 1.0, 8)
noise = np.array([0.12, -0.08, 0.10, -0.12, 0.08, -0.10, 0.07, -0.12])
y_data = f(x_data) + noise

coefficients_degree_7 = np.polyfit(x_data, y_data, 7)
coefficients_degree_3 = np.polyfit(x_data, y_data, 3)

x_inside = np.linspace(-1.0, 1.0, 500)
x_extended = np.linspace(-1.5, 1.5, 700)

y_degree_7_inside = np.polyval(coefficients_degree_7, x_inside)
y_degree_3_inside = np.polyval(coefficients_degree_3, x_inside)
y_degree_7_extended = np.polyval(coefficients_degree_7, x_extended)
y_degree_3_extended = np.polyval(coefficients_degree_3, x_extended)

train_rmse_degree_7 = np.sqrt(np.mean((np.polyval(coefficients_degree_7, x_data) - y_data)**2))
train_rmse_degree_3 = np.sqrt(np.mean((np.polyval(coefficients_degree_3, x_data) - y_data)**2))
inside_rmse_degree_7 = np.sqrt(np.mean((y_degree_7_inside - f(x_inside))**2))
inside_rmse_degree_3 = np.sqrt(np.mean((y_degree_3_inside - f(x_inside))**2))
outside = x_extended > 1.0
outside_rmse_degree_7 = np.sqrt(np.mean((y_degree_7_extended[outside] - f(x_extended[outside]))**2))
outside_rmse_degree_3 = np.sqrt(np.mean((y_degree_3_extended[outside] - f(x_extended[outside]))**2))

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].plot(x_inside, f(x_inside), "k--", label="Underlying function")
axes[0].plot(x_inside, y_degree_7_inside, label="Degree 7 interpolant")
axes[0].plot(x_inside, y_degree_3_inside, label="Degree 3 least-squares fit")
axes[0].scatter(x_data, y_data, color="black", zorder=3, label="Noisy observations")
axes[0].set_title("Within the observed interval")
axes[0].set_xlabel("x")
axes[0].set_ylabel("y")
axes[0].grid(True, alpha=0.3)
axes[0].legend()

axes[1].plot(x_extended, f(x_extended), "k--", label="Underlying function")
axes[1].plot(x_extended, y_degree_7_extended, label="Degree 7 interpolant")
axes[1].plot(x_extended, y_degree_3_extended, label="Degree 3 least-squares fit")
axes[1].scatter(x_data, y_data, color="black", zorder=3, label="Noisy observations")
axes[1].axvline(-1.0, color="grey", linestyle=":")
axes[1].axvline(1.0, color="grey", linestyle=":")
axes[1].set_title("Extrapolation beyond the observed interval")
axes[1].set_xlabel("x")
axes[1].grid(True, alpha=0.3)
axes[1].legend()
fig.tight_layout()
plt.show()

print(f"Degree 7 training RMSE = {train_rmse_degree_7:.6f}")
print(f"Degree 3 training RMSE = {train_rmse_degree_3:.6f}")
print(f"Degree 7 error against f inside [-1, 1] = {inside_rmse_degree_7:.6f}")
print(f"Degree 3 error against f inside [-1, 1] = {inside_rmse_degree_3:.6f}")
print(f"Degree 7 extrapolation error on [1, 1.5] = {outside_rmse_degree_7:.6f}")
print(f"Degree 3 extrapolation error on [1, 1.5] = {outside_rmse_degree_3:.6f}")
```

### Q2C: Why the verification increases confidence

The exact derivative of exp(x) at x = 1 is known to be exp(1), so it provides an independent reference value for checking the numerical implementations. Agreement with this value as the step size is reduced shows that the computed derivatives approach the expected result. In addition, the observed convergence rates agree with the Taylor-series error analysis: approximately first order for the forward difference, second order for the second-order central difference, and fourth order for the fourth-order central difference. This agreement tests both the numerical values and how their errors change with the step size, making many common coding or formula errors more likely to be detected.

This test increases confidence but does not prove that an implementation is correct for every function, evaluation point, or step size. A mistake could go undetected if it happens to cancel for this particular test. It is therefore useful to test additional functions and points, and to examine the behaviour outside the asymptotic range as well.

### Q2D: Asymptotic behaviour and step-size limits

The asymptotic range is the range of step sizes in which the leading Taylor-series truncation error dominates. In this range, halving h reduces the errors by approximately factors of 2, 4, and 16 for the forward, second-order central, and fourth-order central formulas, respectively. The corresponding error curves are close to straight lines on a log-log plot with slopes 1, 2, and 4.

If h is too large, the Taylor expansions are no longer a good local description over the stencil, so the truncation error is large and the derivative approximations become poor. For f(x) = exp(x), the finite-difference estimates grow rapidly as h becomes very large.

In exact arithmetic, reducing h towards zero would make the truncation error tend to zero. In floating-point arithmetic, however, values such as f(x+h) and f(x) become nearly equal. Their subtraction loses significant digits, and division by h amplifies rounding errors. The error therefore eventually stops decreasing and may increase or fluctuate. The plot and selected values demonstrate both limits.

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.exp(x)

def forward_difference(x, h):
    return (f(x + h) - f(x)) / h

def central_difference_2(x, h):
    return (f(x + h) - f(x - h)) / (2.0 * h)

def central_difference_4(x, h):
    return (-f(x + 2.0 * h) + 8.0 * f(x + h) - 8.0 * f(x - h) + f(x - 2.0 * h)) / (12.0 * h)

x0 = 1.0
exact = np.exp(x0)
h_values = np.logspace(-16, 2, 250)
methods = {
    "Forward difference": forward_difference,
    "Second-order central": central_difference_2,
    "Fourth-order central": central_difference_4,
}
errors = {name: np.array([abs(method(x0, h) - exact) for h in h_values]) for name, method in methods.items()}

plt.figure(figsize=(9, 6))
for name, error in errors.items():
    positive = error > 0.0
    plt.loglog(h_values[positive], error[positive], label=name)
plt.axvspan(1e-2, 1e-1, color="grey", alpha=0.15, label="Example truncation-dominated range")
plt.xlabel("Step size h")
plt.ylabel("Absolute error")
plt.title("Finite-difference error from very small to very large step sizes")
plt.grid(True, which="both", alpha=0.3)
plt.legend()
plt.show()

selected_h = [1e-16, 1e-12, 1e-8, 1e-4, 1e-2, 1e-1, 1.0, 10.0, 100.0]
print(" h             Forward error   Central-2 error   Central-4 error")
for h in selected_h:
    values = [abs(method(x0, h) - exact) for method in methods.values()]
    print(f"{h: .1e}   {values[0]: .3e}       {values[1]: .3e}       {values[2]: .3e}")
```

### Q3A: Model selection - logistic population growth

I model the growth of a population living in a closed environment with limited resources. When the population is small compared with the environment's capacity, it grows approximately exponentially. As resources become limited, growth slows and the population approaches a maximum sustainable level.

The model is the logistic ordinary differential equation:

**dP/dt = r P (1 - P/K), with P(0) = P0.**

Here P(t) is the population at time t, r is the intrinsic growth rate, K is the carrying capacity, and P0 is the initial population. I assume the population is well mixed, the environment and parameters remain constant, and there is no migration or random variation. These assumptions make this a simple illustrative model rather than a complete description of a real ecosystem.

I chose this problem because it shows a clear transition from rapid growth to saturation and has a direct interpretation for each parameter. It is also suitable for the verification and validation tasks in this question: the equation has a known analytical solution for checking the numerical code, changing the time step allows a sensitivity study, and synthetic population observations can be compared with model predictions. Large time steps or an incorrect sign in the growth term can also produce clear failure cases.

For the numerical examples, I use dimensionless time and population units, with r = 0.8, K = 100, and P0 = 5. These values are illustrative and are not intended to describe a particular species.

### Q3B: Code verification using the analytical solution

Code verification checks whether the numerical implementation correctly solves the equations it was designed to solve. For the logistic equation, separation of variables gives the analytical solution:

**P(t) = K / [1 + ((K - P0) / P0) exp(-r t)].**

I implement the classical fourth-order Runge-Kutta (RK4) method and compare its numerical solution with this analytical solution at several time steps. A correct RK4 implementation should show an error that decreases approximately in proportion to the fourth power of the time step while the truncation error dominates. I also check the two equilibrium solutions P = 0 and P = K, which should remain constant.

```python
import numpy as np
import matplotlib.pyplot as plt

r = 0.8
K = 100.0
P0 = 5.0
t0 = 0.0
t_final = 10.0

def logistic_rhs(t, P, r, K):
    return r * P * (1.0 - P / K)

def logistic_exact(t, P0, r, K):
    return K / (1.0 + ((K - P0) / P0) * np.exp(-r * t))

def rk4(rhs, t0, y0, t_final, n_steps, args=()):
    h = (t_final - t0) / n_steps
    t = np.linspace(t0, t_final, n_steps + 1)
    y = np.empty(n_steps + 1, dtype=float)
    y[0] = y0
    for i in range(n_steps):
        k1 = rhs(t[i], y[i], *args)
        k2 = rhs(t[i] + h / 2.0, y[i] + h * k1 / 2.0, *args)
        k3 = rhs(t[i] + h / 2.0, y[i] + h * k2 / 2.0, *args)
        k4 = rhs(t[i] + h, y[i] + h * k3, *args)
        y[i + 1] = y[i] + h * (k1 + 2.0 * k2 + 2.0 * k3 + k4) / 6.0
    return t, y

steps_per_interval = np.array([10, 20, 40, 80, 160, 320])
step_sizes = (t_final - t0) / steps_per_interval
maximum_errors = []
for n_steps in steps_per_interval:
    t_num, P_num = rk4(logistic_rhs, t0, P0, t_final, int(n_steps), args=(r, K))
    maximum_errors.append(np.max(np.abs(P_num - logistic_exact(t_num, P0, r, K))))
maximum_errors = np.array(maximum_errors)
observed_orders = np.log(maximum_errors[:-1] / maximum_errors[1:]) / np.log(step_sizes[:-1] / step_sizes[1:])

print("Step size       Maximum absolute error       Observed order")
for i, h in enumerate(step_sizes):
    order = "-" if i == 0 else f"{observed_orders[i - 1]:.3f}"
    print(f"{h: .6f}       {maximum_errors[i]: .6e}             {order}")

t_fine = np.linspace(t0, t_final, 500)
t_plot, P_plot = rk4(logistic_rhs, t0, P0, t_final, 100, args=(r, K))
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].plot(t_fine, logistic_exact(t_fine, P0, r, K), label="Analytical solution")
axes[0].plot(t_plot, P_plot, "o", markevery=8, label="RK4 numerical solution")
axes[0].set_xlabel("Time t")
axes[0].set_ylabel("Population P(t)")
axes[0].set_title("RK4 solution compared with the analytical solution")
axes[0].grid(True, alpha=0.3)
axes[0].legend()
axes[1].loglog(step_sizes, maximum_errors, "o-", label="RK4 maximum error")
reference = maximum_errors[2] * (step_sizes / step_sizes[2])**4
axes[1].loglog(step_sizes, reference, "--", label="Fourth-order reference")
axes[1].set_xlabel("Time step h")
axes[1].set_ylabel("Maximum absolute error")
axes[1].set_title("RK4 convergence against the exact solution")
axes[1].grid(True, which="both", alpha=0.3)
axes[1].legend()
fig.tight_layout()
plt.show()

for equilibrium in [0.0, K]:
    _, equilibrium_solution = rk4(logistic_rhs, t0, equilibrium, t_final, 100, args=(r, K))
    print(f"Maximum change from equilibrium P0 = {equilibrium:g}: {np.max(np.abs(equilibrium_solution - equilibrium)):.3e}")
```

### Q3C: Solution verification by time-step sensitivity

Code verification asks whether the numerical method has been implemented correctly; in Q3B, the RK4 code was compared with the analytical solution. Solution verification asks whether the computed solution for the chosen problem is sufficiently resolved and reliable for its intended use. Here I keep the model and its parameters fixed, reduce the RK4 time step, and compare each trajectory with a much finer-step reference solution.

A user should check that the quantities of interest change by less than an acceptable tolerance as the time step is refined. The solution should also remain physically plausible: for this model, a population starting between zero and K should stay within those bounds and approach K. If the result changes materially under refinement, the time step is too coarse for a reliable conclusion.

```python
import numpy as np
import matplotlib.pyplot as plt

r = 0.8
K = 100.0
P0 = 5.0
t_final = 10.0

def rhs(t, P):
    return r * P * (1.0 - P / K)

def rk4_sensitivity(y0, n_steps):
    h = t_final / n_steps
    t = np.linspace(0.0, t_final, n_steps + 1)
    y = np.empty(n_steps + 1)
    y[0] = y0
    for i in range(n_steps):
        k1 = rhs(t[i], y[i])
        k2 = rhs(t[i] + h / 2.0, y[i] + h * k1 / 2.0)
        k3 = rhs(t[i] + h / 2.0, y[i] + h * k2 / 2.0)
        k4 = rhs(t[i] + h, y[i] + h * k3)
        y[i + 1] = y[i] + h * (k1 + 2.0 * k2 + 2.0 * k3 + k4) / 6.0
    return t, y

step_counts = np.array([25, 50, 100, 200, 400])
reference_steps = 1600
t_reference, P_reference = rk4_sensitivity(P0, reference_steps)
step_sizes = t_final / step_counts
max_differences = []
end_differences = []
trajectories = []

for n_steps in step_counts:
    t, P = rk4_sensitivity(P0, int(n_steps))
    reference_indices = np.arange(n_steps + 1) * (reference_steps // n_steps)
    P_ref_on_grid = P_reference[reference_indices]
    max_differences.append(np.max(np.abs(P - P_ref_on_grid)))
    end_differences.append(abs(P[-1] - P_ref_on_grid[-1]))
    trajectories.append((t, P))

max_differences = np.array(max_differences)
end_differences = np.array(end_differences)
print("Step size    Maximum trajectory difference    Final-time difference    Population range")
for h, n_steps, max_diff, end_diff, (_, P) in zip(step_sizes, step_counts, max_differences, end_differences, trajectories):
    print(f"{h: .4f}       {max_diff: .6e}                 {end_diff: .6e}        [{P.min():.3f}, {P.max():.3f}]")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for h, (t, P) in zip(step_sizes, trajectories):
    axes[0].plot(t, P, label=f"h = {h:g}")
axes[0].plot(t_reference, P_reference, "k--", linewidth=2, label="Fine-step reference")
axes[0].set_xlabel("Time t")
axes[0].set_ylabel("Population P(t)")
axes[0].set_title("Effect of time-step refinement")
axes[0].grid(True, alpha=0.3)
axes[0].legend()
axes[1].loglog(step_sizes, max_differences, "o-")
axes[1].set_xlabel("Time step h")
axes[1].set_ylabel("Maximum difference from reference")
axes[1].set_title("Sensitivity of the computed trajectory")
axes[1].grid(True, which="both", alpha=0.3)
fig.tight_layout()
plt.show()
```

### Q3D: Model validation with synthetic observations

In a real study, validation would require repeated population counts over time, an estimate of the initial population, and information about changes in habitat, resources, births, deaths, and migration. The data should cover enough of the growth curve to assess both the initial growth rate and the approach to carrying capacity. Measurement uncertainty and possible changes in environmental conditions should also be documented.

Here I demonstrate the comparison process with synthetic observations generated from a logistic population trajectory and perturbed by reproducible measurement noise. I estimate the growth rate and carrying capacity using nonlinear least squares with SciPy's least_squares function, which minimizes the sum of squared differences between the model and the calibration observations. I then compare the resulting RK4 prediction with held-out observations that were not used for parameter estimation.

The fitted parameters are close to the values used to generate the synthetic data, and the held-out RMSE is 1.734 population units, comparable to the 1.5-unit standard deviation of the added measurement noise. This indicates reasonable agreement for this synthetic example.

Because the synthetic observations are generated from the same logistic model, this example checks the validation workflow rather than establishing that the logistic model describes a real population. Real validation would require independent field observations and would also need to assess whether the model assumptions are appropriate.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import least_squares

P0 = 5.0
r_true = 0.8
K_true = 100.0

def logistic_exact(t, r, K, P0):
    return K / (1.0 + ((K - P0) / P0) * np.exp(-r * t))

rng = np.random.default_rng(42)
t_observed = np.arange(0.0, 11.0, 1.0)
P_true = logistic_exact(t_observed, r_true, K_true, P0)
P_observed = P_true + rng.normal(0.0, 1.5, size=t_observed.size)
calibration_indices = np.arange(0, t_observed.size, 2)
validation_indices = np.arange(1, t_observed.size, 2)
t_calibration = t_observed[calibration_indices]
P_calibration = P_observed[calibration_indices]
t_validation = t_observed[validation_indices]
P_validation = P_observed[validation_indices]

def calibration_residuals(parameters):
    r, K = parameters
    return logistic_exact(t_calibration, r, K, P0) - P_calibration

fit = least_squares(calibration_residuals, x0=np.array([0.6, 110.0]), bounds=([0.01, P0 + 0.01], [2.0, 300.0]))
r_estimated, K_estimated = fit.x

def logistic_rhs(t, P, r, K):
    return r * P * (1.0 - P / K)

def rk4_validation(y0, t_final, n_steps, r, K):
    h = t_final / n_steps
    t = np.linspace(0.0, t_final, n_steps + 1)
    y = np.empty(n_steps + 1)
    y[0] = y0
    for i in range(n_steps):
        k1 = logistic_rhs(t[i], y[i], r, K)
        k2 = logistic_rhs(t[i] + h / 2.0, y[i] + h * k1 / 2.0, r, K)
        k3 = logistic_rhs(t[i] + h / 2.0, y[i] + h * k2 / 2.0, r, K)
        k4 = logistic_rhs(t[i] + h, y[i] + h * k3, r, K)
        y[i + 1] = y[i] + h * (k1 + 2.0 * k2 + 2.0 * k3 + k4) / 6.0
    return t, y

t_model, P_model = rk4_validation(P0, 10.0, 1000, r_estimated, K_estimated)
validation_indices_on_grid = np.rint(t_validation / (10.0 / 1000)).astype(int)
P_validation_prediction = P_model[validation_indices_on_grid]
calibration_rmse = np.sqrt(np.mean(calibration_residuals(fit.x)**2))
validation_rmse = np.sqrt(np.mean((P_validation_prediction - P_validation)**2))

print(f"True synthetic parameters: r = {r_true:.3f}, K = {K_true:.3f}")
print(f"Estimated parameters: r = {r_estimated:.3f}, K = {K_estimated:.3f}")
print(f"Calibration RMSE = {calibration_rmse:.3f} population units")
print(f"Held-out validation RMSE = {validation_rmse:.3f} population units")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].plot(t_model, P_model, label="RK4 model prediction")
axes[0].plot(t_model, logistic_exact(t_model, r_true, K_true, P0), "k--", label="Synthetic generating curve")
axes[0].scatter(t_calibration, P_calibration, color="tab:blue", zorder=3, label="Calibration observations")
axes[0].scatter(t_validation, P_validation, color="tab:orange", marker="s", zorder=3, label="Held-out observations")
axes[0].set_xlabel("Time t")
axes[0].set_ylabel("Population")
axes[0].set_title("Logistic model compared with synthetic data")
axes[0].grid(True, alpha=0.3)
axes[0].legend()
axes[1].axhline(0.0, color="black", linewidth=1)
axes[1].scatter(t_validation, P_validation_prediction - P_validation, color="tab:orange")
axes[1].set_xlabel("Validation time t")
axes[1].set_ylabel("Prediction - observation")
axes[1].set_title("Held-out validation residuals")
axes[1].grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

### Q3E: Failure demonstrations

The first failure uses the explicit Euler method with a time step that is too large for the population dynamics. The Euler update can overshoot the carrying capacity and produce a negative population, which is physically impossible. The second failure is a coding error: I change the minus sign in the logistic growth term to a plus sign. This removes the limiting effect of resources and allows the computed population to exceed the carrying capacity.

In both plots the failed calculation is compared with a fine-step RK4 solution of the intended model. These examples distinguish a poor numerical choice from an incorrect implementation: reducing the Euler time step addresses the first failure, while the sign error requires correcting the equation in the code.

```python
import numpy as np
import matplotlib.pyplot as plt

r = 0.8
K = 100.0
P0 = 5.0

def correct_rhs(t, P):
    return r * P * (1.0 - P / K)

def faulty_rhs(t, P):
    return r * P * (1.0 + P / K)

def rk4_failure_demo(rhs, t_final, n_steps):
    h = t_final / n_steps
    t = np.linspace(0.0, t_final, n_steps + 1)
    y = np.empty(n_steps + 1)
    y[0] = P0
    for i in range(n_steps):
        k1 = rhs(t[i], y[i])
        k2 = rhs(t[i] + h / 2.0, y[i] + h * k1 / 2.0)
        k3 = rhs(t[i] + h / 2.0, y[i] + h * k2 / 2.0)
        k4 = rhs(t[i] + h, y[i] + h * k3)
        y[i + 1] = y[i] + h * (k1 + 2.0 * k2 + 2.0 * k3 + k4) / 6.0
    return t, y

t_euler = np.arange(0.0, 30.0 + 5.0, 5.0)
P_euler = np.empty(t_euler.size)
P_euler[0] = P0
for i in range(t_euler.size - 1):
    h = t_euler[i + 1] - t_euler[i]
    P_euler[i + 1] = P_euler[i] + h * correct_rhs(t_euler[i], P_euler[i])

t_reference, P_reference = rk4_failure_demo(correct_rhs, 30.0, 3000)
t_bug, P_bug = rk4_failure_demo(faulty_rhs, 3.0, 300)
crossing = np.where(P_bug > K)[0]

print("Coarse Euler results:")
for t, P in zip(t_euler, P_euler):
    print(f"t = {t: .1f}, P = {P: .6f}")
print(f"Minimum coarse Euler population = {P_euler.min():.6f}")
if crossing.size:
    print(f"Wrong-sign model first exceeds K at t = {t_bug[crossing[0]]:.2f}")
print(f"Wrong-sign model population at t = 3 = {P_bug[-1]:.6f}")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].plot(t_reference, P_reference, label="Correct model, fine-step RK4")
axes[0].plot(t_euler, P_euler, "o-", label="Explicit Euler with h = 5")
axes[0].axhline(0.0, color="black", linewidth=1)
axes[0].axhline(K, color="grey", linestyle=":", label="Carrying capacity K")
axes[0].set_xlabel("Time t")
axes[0].set_ylabel("Population P(t)")
axes[0].set_title("Failure from an excessively large Euler step")
axes[0].grid(True, alpha=0.3)
axes[0].legend()
axes[1].plot(t_reference[t_reference <= 3.0], P_reference[t_reference <= 3.0], label="Correct model, fine-step RK4")
axes[1].plot(t_bug, P_bug, label="Wrong-sign model")
axes[1].axhline(K, color="grey", linestyle=":", label="Carrying capacity K")
axes[1].set_xlabel("Time t")
axes[1].set_ylabel("Population P(t)")
axes[1].set_title("Failure from an incorrect growth-term sign")
axes[1].grid(True, alpha=0.3)
axes[1].legend()
fig.tight_layout()
plt.show()
```

### Q3F: Accuracy, efficiency and stability

Accuracy describes how close the computed population trajectory is to the exact logistic solution. It can be assessed using the maximum absolute error over the simulated time interval.

Efficiency describes the computational work required to achieve a chosen accuracy. Here I compare Euler and RK4 by counting evaluations of the model right-hand side needed to reach the same error tolerance. This count is a reproducible measure of work: Euler uses one evaluation per step, while classical RK4 uses four. In practice, the best method depends on the tolerance and the cost of each evaluation.

Stability describes whether numerical errors and perturbations remain bounded rather than growing into an unphysical solution. For this population model, a numerical solution starting between zero and K should remain physically plausible. The Q3E Euler calculation with a very large step produced a negative population, demonstrating a stability failure for that step size.

```python
import numpy as np
import matplotlib.pyplot as plt

r = 0.8
K = 100.0
P0 = 5.0
t_final = 10.0

def exact_solution(t):
    return K / (1.0 + ((K - P0) / P0) * np.exp(-r * t))

def rhs(P):
    return r * P * (1.0 - P / K)

def solve_euler(n_steps):
    h = t_final / n_steps
    y = np.empty(n_steps + 1)
    y[0] = P0
    for i in range(n_steps):
        y[i + 1] = y[i] + h * rhs(y[i])
    t = np.linspace(0.0, t_final, n_steps + 1)
    return t, y

def solve_rk4(n_steps):
    h = t_final / n_steps
    t = np.linspace(0.0, t_final, n_steps + 1)
    y = np.empty(n_steps + 1)
    y[0] = P0
    for i in range(n_steps):
        k1 = rhs(y[i])
        k2 = rhs(y[i] + h * k1 / 2.0)
        k3 = rhs(y[i] + h * k2 / 2.0)
        k4 = rhs(y[i] + h * k3)
        y[i + 1] = y[i] + h * (k1 + 2.0 * k2 + 2.0 * k3 + k4) / 6.0
    return t, y

comparison_steps = np.array([10, 20, 40, 80, 160])
comparison_sizes = t_final / comparison_steps
euler_errors = []
rk4_errors = []
for n_steps in comparison_steps:
    t_euler, P_euler = solve_euler(int(n_steps))
    t_rk4, P_rk4 = solve_rk4(int(n_steps))
    euler_errors.append(np.max(np.abs(P_euler - exact_solution(t_euler))))
    rk4_errors.append(np.max(np.abs(P_rk4 - exact_solution(t_rk4))))

tolerance = 5e-2
candidate_steps = [10, 20, 40, 80, 160, 320, 640, 1280, 2560, 5120, 10240]
costs = {}
for name, solver, evaluations_per_step in [("Euler", solve_euler, 1), ("RK4", solve_rk4, 4)]:
    for n_steps in candidate_steps:
        t, P = solver(n_steps)
        error = np.max(np.abs(P - exact_solution(t)))
        if error <= tolerance:
            costs[name] = (n_steps, evaluations_per_step * n_steps, error)
            break

print("Method   Steps   RHS evaluations   Maximum error")
for name, (n_steps, evaluations, error) in costs.items():
    print(f"{name:<7} {n_steps:>6} {evaluations:>16}   {error:.3e}")

coarse_h = 5.0
coarse_euler = [P0]
for _ in range(6):
    coarse_euler.append(coarse_euler[-1] + coarse_h * rhs(coarse_euler[-1]))
print(f"Euler with h = 5 gives minimum population {min(coarse_euler):.3f}; the physical range is [0, {K:g}].")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].loglog(comparison_sizes, euler_errors, "o-", label="Euler")
axes[0].loglog(comparison_sizes, rk4_errors, "o-", label="RK4")
axes[0].set_xlabel("Time step h")
axes[0].set_ylabel("Maximum absolute error")
axes[0].set_title("Accuracy against the exact solution")
axes[0].grid(True, which="both", alpha=0.3)
axes[0].legend()
axes[1].bar(list(costs.keys()), [entry[1] for entry in costs.values()], color=["tab:blue", "tab:orange"])
axes[1].set_yscale("log")
axes[1].set_ylabel("Right-hand-side evaluations")
axes[1].set_title(f"Work to reach maximum error <= {tolerance:g}")
axes[1].grid(True, axis="y", which="both", alpha=0.3)
fig.tight_layout()
plt.show()
```
