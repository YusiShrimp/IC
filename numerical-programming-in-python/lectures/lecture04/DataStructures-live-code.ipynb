```python
class myclass():

    def __init__(self, val):
        self.val = val

    def __str__(self):
        return f"in __str__ with value {self.val}"

    def __call__(self):
        print(f"in __call__ with value {self.val}")

    def __repr__(self):
        return f"in __repr__ with value {self.val}"

    def __add__(self, other):
        newcls = myclass(self.val + other.val)
        return newcls

    def __sub__(self, other):
        newcls = myclass(self.val - other.val)
        return newcls

    def __mul__(self, other):
        newcls = myclass(self.val * other.val)
        return newcls

    def __truediv__(self, other):
        newcls = myclass(self.val / other.val)
        return newcls

    def __floordiv__(self, other):
        newcls = myclass(self.val // other.val)
        return newcls
    
```

```python
inst1 = myclass(1)
```

```python
print(inst1)
```

```python
inst1()
```

```python
inst1
```
