## Integers

It may be a surprise, but the built-in Python integers (& floats & strings) are immutable. 

To "update" an integer, Python creates a new object with the updated value.  We can check this using the `id()` function.

This returns a unique value (actually an integer) for every Python object in memory.

````python
x = 3
print(f'1: id of object named x is {id(x)}')
x += 5
print(f'2: id of object named x is {id(x)}')
````

For efficiency, Python actually stores a single set of Python objects representing the small integers (from -5 to 256).

Remember we can use the `is` operator to check if two variable names point to the same object.

````python
x = 3; y = 3
print(f'ids: {id(x)}, {id(y)} | Same object: {x is y}')

a = 12345; b = 12345
print(f'ids: {id(a)}, {id(b)} | Same object: {a is b}')    
````

As we heard on Monday, Python has "arbitrary length precision" integers.

Fundamental limit on the largest integer number your Python code can work with is the memory available in your computer.

````python
import sys

a = 10**0
print(f'a uses {sys.getsizeof(a)} bytes')
b = 10**5
print(f'b uses {sys.getsizeof(b)} bytes')
c = 10**10
print(f'c uses {sys.getsizeof(c)} bytes')
d = 10**1000
print(f'd uses {sys.getsizeof(d)} bytes')
````

## Lists

Python lists are formed using `[]` brackets, or using the `list()` function on another collection or iterator.


```python
list_a = [1, 3, 5]
list_b = list((3, 4, 5))
list_c = [ _ for _ in range(6, 10) ]
```

Python lists are:
- mutable
- one-dimensional (i.e. single index)
- ordered

_collections_ of other Python objects.

They also support _iteration_ over the index.

Lists are (relatively) fast at:
- finding/changing their elements,
- adding/deleting the _last_ element in the list.

They are much slower at adding or removing elements other than the last one. This scales with the length of the list

### List comprehension

The pattern `[ _ for _ in range(6, 10) ]` is called a list comprehension:
- _Usually_ the fastest way to build a list in Python.
- Good idea to use as long as it stays readable.

Let's try timing two methods for creating a list, to see which is faster

````python
def list_method_one():
    out = []
    for i in range(10000):
        out.append(i)
    return out

%timeit a = list_method_one()
````

````python
def list_method_two():
    return [i for i in range(10000)]

%timeit b = list_method_two()
````

Notice that method two (the list comprehension) runs faster than the loop in method one (by about 20–25% on my computer).

### Implementation

In CPython, (i.e the Python on most computers) lists are a block of memory holding a sequence of "memory addresses".

|list index| address|
|-|-|
|0|#2343|
|1|#323|
|2|#12|
|3|#323|
|4| _not used yet_|
|5| _not used yet_|
|6| _not used yet_|
|7| _not used yet_|

Looking up the `n`th element is quick (jump straight to the nth block, look up the contents of the address).

Implementations grab spare memory to allow room to expand. Adding a new element (`x.append()`) is normally cheap unless it's "full".

Adding to a full list requires copying the whole list. Deleting & inserting elements is also expensive

|index| _original_ | _append_ | _insert_ | _remove_ |
|-|-|-|-|-|

|6| _not used yet_|#226|#67| _not used yet_|

## Tuples

Python tuples are formed using commas (usually in `()` brackets) or using the `tuple()` function on another collection or iterator.


```python
tuple_a = 1, 3, 5
tuple_b = (2, 3, 4)
tuple_c = tuple([3, 4, 5])
```

Python tuples are:
- immutable
- one-dimensional
- ordered

_collections_ of other Python objects.

Tuples are the immutable equivalent of lists.

Tuples are fast at:
-  finding their elements

They tend to use slightly less memory than a list.

If you know that something should never change, then use a tuple.

### Implementation

In CPython, tuples are implemented a lot like lists.

Block can be chosen to be exactly the right size to hold all the addresses. No need for room to expand.

### Tuple unpacking

A useful feature of tuples is that they can be unpacked into multiple variables in a single statement. For example, you can write 

```python
a, b, c = 1, 2, 3
```

to assign the integers 1, 2, and 3 to the variable names `a`, `b` and `c`.

You often see this feature used to collect the values returned from a function.

## Generators

Unlike lists, you can't build a tuple using a comprehension (tuple comprehension doesn't exist). 

But there is a similar syntax which might look like what you would expect a tuple comprehension to look like; these are called **generators**.

````python
for y in (x**2 for x in range(3)):
    print(y)
````

Generators can only be iterated over once. They do not store their contents, but instead generate the relevant values as they are needed, and then discard them once they've been used.

They are therefore useful as they are memory efficient.

However, can lead to confusion:

````python
gen = (x for x in range(4))
print(list(gen))
print(list(gen))
````

## Sets

Python sets are formed using  `{}` brackets or using the `set()` function on another collection or iterator.

```python
set_a = {1, 3, 5}
set_b = set([3, 4, 5])
```

Python sets are:
- unordered
- mutable

_collections_ of **distinct** objects.

It is quick to test if a value is in a set using the `x in S` syntax.

Let's compare how long it takes to check if a value is in a set compared to a list of the same length.

````python
import random

x = 12345
S = {random.randint(0, 1_000_000) for _ in range(1000)}
L = list(S)
%timeit x in S
%timeit x in L
````

So it's a couple of orders of magnitude quicker to test if something is in the set compared to the list.

### Implementation

CPython sets are implemented via a "hash table".

This consists of:
- a large (mostly empty) block of memory, of length N
- a "hash" function which maps inputs to the numbers 0 to N-1.

|index| address|
|-|-|
|0| value1|
|1| _not used yet_|
|2| value3|
|3| _not used yet_|
|4| _not used yet_|
|5| _not used yet_|
|6| value2|
|7| _not used yet_|

To add a member:
- calculate hash of value
- insert in that space

To check if a value is in the set:
- calculate hash of value
- check that space

This is all relatively fast.

However, you may have spotted that it's possible a hash function will give the same location for two different inputs.

We need a process to deal with these "collisions".

Python's method is essentially to carry on looking in other blocks via a predictable pattern based on the hash until:
- it finds what it's looking for, or
- it finds an empty entry.

This is why sets need to be quite empty, using more memory than other data types like lists.

## Frozensets

Python frozensets are formed by using the `frozenset()` function on another collection or iterator.

They are the immutable equivalent of a set (i.e., their contents cannot be changed after they have been created).

## Dictionaries (dicts)

Python dictionaries are mapping objects that map _keys_ to arbitrary _values_. 

They are formed using `{}` brackets on comma separated, colon delimited pairs of objects. Alternatively, the `dict()` function can also be used to create a new dictionary, or you can use a dict comprehension.

```python
dict_a = {"a": 1, "b": 10, "c": 25}
dict_b = dict(x=1, y=2, z=100)
dict_c = {x: x ** 2 for x in range(10)}
```


Like sets, dicts are optimized to test if a key is in it. Keys must be _immutable_ (e.g. no lists) and _hashable_.

````python
def num(maxval=1_000_000):
    return random.randint(0, maxval)

# make a dictionary of random numbers using a dict comprehension
D = {num():num() for _ in range(1000)}

# get the equivalent list
L = list(D.values())

x = 12345
%timeit x in D
%timeit x in L 
````

Again, it's a couple of orders of magnitude quicker to check if a key is in the dict compared to the list

Dictionary comprehensions are useful to build a dict quickly, but not always a speed benefit.

````python
# Reversing a dictionary via a "for" loop versus
# a dict comprehension

def method_one(input):
    out = {}
    for key, val in input.items():
        out[val] = key
    return out

def method_two(input):
    return {val:key for key, val in input.items()}
````

````python
test = {_:20000-_ for _ in range(10000)}
%timeit t1 = method_one(test)  # for loop
%timeit t2 = method_two(test)  # comprehension
````

Both methods take around the same time in this case.

### Implementation

CPython implements dicts using hash tables much like sets. 

There are some interesting differences between sets and dicts:

* a `dict` stores both a note of the key & (the address of) the object storing value
* a `dict` stores the insertion order (in Python version 3.7+), whereas a `set` does not

# Part III: Python Classes

Python allows programmers to "build your own" data structures called _classes_. Follows a world-view called "object oriented programming".

Classes allow you to create your own holders for data,  plus functions (class methods) which you need to modify it.

As you saw in the presessional Python material, a class (say, `MyClass`) is defined using a syntax like this:

````python
class MyClass:
    """
    This is my cool class.
    """

    def my_method(self, a, b):
        """ Add two values, store the result, then return the product"""
        self.val = a + b
        return a * b
````

We can then create instances of `MyClass` using the `()` operator:

````python
# Create an instance of the class
inst = MyClass()
````

````python
a = inst.my_method(1, 2)
print(f'inst.val: {inst.val} | value returned from method: {a}')
````

## Why use classes at all?

Classes (and class objects) are not a required part of a programming language, and many languages using other programming paradigms (for example, purely procedural languages such as C) have existed for a long time. 

However, an object-oriented approach works well at avoiding coding mistakes when combining structured collections of data and functions modifying or processing that data. It does this by binding the data and the method together, making it much harder to apply them to the wrong thing (Encapsulation).

## The `__init__` method

We can use the special `__init__()` method to _initialize_ instance variables when the new object is created. For example, say we want to store some coordinates in a `Point3D` class:

````python
class Point3D:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z
        
pnt1 = Point3D(1, 4, 7)

print(pnt1.x, pnt1.y, pnt1.z)
````

## Simple class example: Trees
Here's a simple example of a Python class, in which:
* Each instance of the class stores some information about a tree (species, age, height)
* There are some methods which interact with that data
* We can easily write code to access the data from these instances

````python
class Tree:
    def __init__(self, species, age, height):
        self.species = species
        self.age = age  # age in years
        self.height = height  # height in metres

    def __str__(self):
        return f"{self.species} tree is {self.age} years old, {self.height} metres tall"
    
    def estimate_canopy_diameter(self):
        # Simplified estimation: canopy diameter is roughly
        # 1/3 of the height plus 0.1 metres per year of age
        return (self.height / 3) + (0.1 * self.age)
````

````python
tree1 = Tree("Oak", 30, 25)
tree2 = Tree("Sycamore", 75, 15)
tree3 = Tree("Birch", 5, 8)
forest = [tree1, tree2, tree3]

for tree in forest:
    print(tree)
    print(f"{tree.species} canopy diameter = {tree.estimate_canopy_diameter(): .2f} m")
    print()
````

We can also perform operations on the underlying data; e.g. here we use the `max` function:

````python
# Find the tallest tree in the forest
tallest = max(forest, key=lambda tree: tree.height)
print(f"The tallest tree is the {tallest.species} tree, which is {tallest.height} m tall.")

# Find the widest tree in the forest
widest = max(forest, key=lambda tree: tree.estimate_canopy_diameter())
print(f"The widest tree is the {widest.species} tree, which is {widest.estimate_canopy_diameter(): .2f} m wide.")
````

That was just a toy example. Often classes will become a lot more complicated than this, and have many more methods and variables.

## Class attributes vs Instance attributes

A common pitfall is defining things at the _class_ level rather than the _instance_ level. For example, using our `MyClass` from earlier, we could add a list (`mylist`) at the class level like this:

````python
class MyClass:
    """
    This is my cool class.
    """

    # This is the class level
    # Note, it is not defined inside a function like self.val below
    mylist = []

    def my_method(self, a, b):
        """ Add two values, store the result, then return the product"""
        self.val = a + b
        return a*b
````

Note that attributes at class level are the _same_ object for all instances (e.g. `mylist` in the example above).

````python
# Get two instances
inst1 = MyClass()
inst2 = MyClass()
````

````python
# Attributes created on the class are the same variable
inst1.mylist.append(1)
inst2.mylist.append(2)

print(f'inst1.mylist is inst2.mylist : {inst1.mylist is inst2.mylist}')
print(f'inst1.mylist: {inst1.mylist}')
print(f'inst2.mylist: {inst2.mylist}')
````

Attributes set at the instance level (e.g. `val`) can differ from each other:

````python
# Attributes created on instances can differ
inst1.my_method(1, 2)
inst2.my_method(4, 5)

print(f'inst1.val is inst2.val: {inst1.val is inst2.val}')

print(f'inst1.val: {inst1.val}')
print(f'inst2.val: {inst2.val}')
````

## Inheritance

Python classes can inherit features (methods and attributes) from other classes. Remember from Part I this morning that all Python data is an "object". Every class, from `int` to `list` and everything else we have seen and written, ultimately inherits from the base `object` class.

By default, new classes will inherit from the base `object` class. We can instead take any python class and modify it to suit our needs.

We do this using the following syntax:

```python
class MySubClass(MyBaseClass):
    '''MySubClass will inherit from MyBaseClass'''
    
    # define any new methods here
```

In our `Tree` example, we could make a subclass to inherit some attributes and methods from the base class. 

For example, here we will make a `Shrub` subclass, which inherits the `__init__` method and the `species`, `age`, and `height` attributes from `Tree`.

We then redefine the `__str__` and `estimate_canopy_diameter` methods with some different functionality.

````python
class Shrub(Tree):
    '''Shrub will inherit from our Tree class'''
    
    def __str__(self):
        ''' Here we redefine the __str__ function to change tree to shrub '''
        return f"{self.species} shrub is {self.age} years old, {self.height} metres tall"
    
    def estimate_canopy_diameter(self):
        return (self.height / 5) + (0.2 * self.age)
````

````python
rose = Shrub("Rose", 5, 1.3)
print(rose)
print(rose.estimate_canopy_diameter())
````

To check what the base class of a subclass is, we can use the special `.__base__` attribute:

````python
print(f"The base class of Tree is {Tree.__base__}")
print(f"The base class of Shrub is {Shrub.__base__}")
````

### Modifying an inherited function without rewriting the whole thing
Sometimes we might want a function to do most of what the base class does, but add some extra functionality too. We can make use of `super()` here to access the functions in the base class.

For example, let's say we wanted to just add a new attribute to our `Shrub` class in `__init__` that was not present in `Tree`.

````python
class Shrub(Tree):
    '''Shrub will inherit from our Tree class'''

    def __init__(self, species, age, height, fruit):
        super().__init__(species, age, height)
        self.fruit = fruit
    
    def __str__(self):
        ''' Here we redefine the __str__ function to change tree to shrub '''
        fruitstr = "It bears fruit" if self.fruit else "It does not bear fruit"
        return f"{self.species} shrub is {self.age} years old, {self.height} m tall. {fruitstr}."
    
    def estimate_canopy_diameter(self):
        return (self.height / 5) + (0.2 * self.age)
````

````python
rose = Shrub("Rose", 5, 1.3, False)
print(rose)

blackberry = Shrub("Blackberry", 3, 2, True)
print(blackberry)
````

## Properties: combining class methods and attributes

[*Note: This is quite advanced usage, but is still useful to know about*]

Sometimes we want a quantity which "feels" like data (i.e. _attribute_) but is set by a function.

We can do this with the Python _decorators_: `@property` and `@<attribute>.setter`:

````python
class Temperature:
    
    def __init__(self, temp):
        self.celsius = temp
       
    @property
    def fahrenheit(self):
        return self.celsius * 9 / 5 + 32
     
    @fahrenheit.setter
    def fahrenheit(self, temp):
        self.celsius = (temp - 32) * 5 / 9
````

Note how `fahrenheit` is calculated from the `celsius` attribute, and we define a separate _setter_ method, which updates the `celsius` value.

````python
t1 = Temperature(30.)
    
print(t1.celsius, t1.fahrenheit)

t1.celsius = 20.

print(t1.celsius, t1.fahrenheit)

t1.fahrenheit = 50.

print(t1.celsius, t1.fahrenheit)
````

## A note on `__init__` and `__new__`

[*Note: This is advanced usage--can safely ignore this for now*]

As always things are actually a little more complicated.
When a new instance is created, Python calls the `__new__` class method, and then calls `__init__` on the object that `__new__` returns.

Generally, this is what we want to happen, so that we can ignore `__new__` (i.e., use the `__new__` function from our base class), and just write the `__init__` method.

However, very rarely, you may want to write a `__new__` function to do something special.

Let's see an example of how to do that.

````python
#An example of a class with a __new__ method

import weakref

point_cache = []

class Point3D:
    
    def __new__(cls, *args):
        print('in __new__')
        # call the original object creation method
        point = super().__new__(cls) 
        point_cache.append(weakref.proxy(point))
        return point   # important!
    
    def __init__(self, x, y, z):
        print('in __init__')
        self.x = x 
        self.y = y
        self.z = z
        
    def __del__(self):
        point_cache.remove(self)
        
    def __repr__(self):
        return f'Point3D({self.x}, {self.y}, {self.z})'
````

````python
point1 = Point3D(1, 1, 1)
print(point_cache)
del point1
print(point_cache)
````

As a note, the opposite of the `__new__` method is the `__del__` method. This gets called at the point the Python garbage collector believes the object is no longer needed, so the `__del__` method can be used to clean up anything special.
