### Linear algebra with numpy.linalg
 
Numpy provides some linear algebra capabilities, from matrix-vector product to matrix inversion and system solution

Simple matrix vector product within numpy core:

````python
import numpy as np

A = np.array([[1, 2, 3],
              [4, 5, 6],
              [7, 8, 8]])
B = np.array([1, 2, 1])
````

````python
# This doesn't give us the matrix vector product!
print(A * B)
````

````python
# All of these methods work here
print(f"np.dot     : {np.dot(A, B)}")
print(f"@ operator : {A @ B}")
print(f"np.matmul  : {np.matmul(A, B)}")
````

A quick note about these different methods. 

- The `@` operator and `np.matmul` are functionally equivalent, and will give the same result for any dimension of array. 

- `np.dot` works in this case, but for higher-dimension arrays (i.e. higher than 2D), will do something a little different — it performs a sum product over specific axes rather than true matrix multiplication. 

Try looking at the [`np.dot`](https://numpy.org/doc/stable/reference/generated/numpy.dot.html) and [`np.matmul`](https://numpy.org/doc/stable/reference/generated/numpy.matmul.html#numpy.matmul) documentation for more.

````python
# example from the np.matmul documentation
a = np.ones([9, 5, 7, 4])  
c = np.ones([9, 5, 4, 3])
print(np.dot(a, c).shape)
print(np.matmul(a, c).shape)
# n is 7, k is 4, m is 3
````

  The `numpy.linalg` module is necessary for more complex operations.

````python
import numpy.linalg as la
````

Key matrix/vector quantities can be computed from arrays:

````python
# Let's look at the documentation to see what this does!
la.norm?
````

````python
n = la.norm(B)
print(n)

n = la.norm(A)
print(n)

d = la.det(A)
print(d)
````

It is possible to solve linear systems, using fast, low level C/Fortran code:

````python
x = la.solve(A, B)  # find x for Ax = B
print(x)
````

or to invert matrices (generally not a good idea in practice, large computational cost, and often inaccurate!)

````python
A_inv = la.inv(A)
print(A_inv)
````

In terms of serious mathematics, complex quantities like the eigen decomposition (of a square matrix) can also be computed:

````python
eival, eivec = la.eig(A)

print(f"Eigenvalues: {eival}")

print("Eigenvectors:")
print(eivec)

np.isclose(A @ eivec[:, 0], eival[0] * eivec[:, 0])  # Av = λv
````

### Optimisation

`scipy.optimise` provides a very wide range of optimisation methods: Newton-like minimization algorithms, least square methods, root finding... 

NB. Examples below are inspired from the official documentation. Interesting examples require setting up complex problems and exceed the scope of this presentation.

````python
import scipy.optimize as sopt
````

Let's find the minimum of a simple 1D function: <br>

$$ f(x) = -e^{-(x-0.2)^2} $$

````python
## let's define a simple function

def f(x):
    return -np.exp(-(x - 0.2)**2)

result = sopt.minimize_scalar(f)

x_min = result.x
print(x_min)
````

And now a more complex 2D one:

$$ f(x,y)=\frac{1}{2}(1-x)^{2}+(y-x^{2})^{2} $$

````python
def f(x):
    """Rosenbrock function"""
    return .5*(1 - x[0])**2 + (x[1] - x[0]**2)**2

def jacobian(x):
    """Partial derivatives of the Rosenbrock function"""
    return np.array((-2*.5*(1 - x[0]) - 4*x[0]*(x[1] - x[0]**2), 2*(x[1] - x[0]**2)))

sopt.minimize(f, x0=[2, -1], method="Newton-CG", jac=jacobian) 
````

### Basic plots


Let's prepare some data:

````python
import numpy as np
%matplotlib widget
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)
z = np.sin(x+np.pi/2)
````

Plotting this data is as simple as:

````python
fig = plt.figure()

plt.plot(x, y)
plt.plot(x, z)
plt.show()
````

Alternative approach, plotting multiple lines with one call to `plt.plot`

````python
fig = plt.figure()

plt.plot(x, y, x, z)
plt.show()
````

Generally recommended to use `matplotlib`'s object-oriented interface. Here we create the same plot, but first creating a figure and axes instance, and then calling methods from those instances.

````python
fig = plt.figure()
ax = fig.add_subplot()  # Add a subplot (Axes) to a Figure

ax.plot(x, y)  # Using the plot method from the Axes instance
ax.plot(x, z)
plt.show()
````

We can specify axes limits and titles, and a global plot title:

````python
fig = plt.figure()
ax = fig.add_subplot()

ax.set_xlim((x.max()*0.25, x.max()*0.75))
ax.set_xlabel("x")
ax.set_ylabel("my y label")
ax.set_title("Trigonometry")
ax.plot(x, y, x, z)
plt.show()
````

as well as legends:

````python
fig, ax = plt.subplots()  # we can create fig and ax on one line!

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.plot(x, y, x, z)

# Add a legend
ax.legend(("sin(x)", "sin(x+pi/2)"))
plt.show()
````

Or keep the labels together with the plot command, so you don't accidently label the wrong line!

````python
fig, ax = plt.subplots()  # we can create fig and ax on one line!

ax.set_xlabel("x")
ax.set_ylabel("y")

# use the label kwarg
ax.plot(x, y, label="sin(x)")
ax.plot(x, z, label="sin(x+pi/2)")

# Add a legend
ax.legend(loc=2)
plt.show()
````

### Subplots

Sometimes, you will want to group several plots in one figure. Matplotlib provides an flexible way of managing subplots.
```
    (fig, axes) = plt.subplots(nrows, ncols)
```
`subplots()` returns a tuple with a reference to the figure (`fig`) and an array of references to each subplot axes (`axes`). In the cells above, we've been using `plt.subplots` with it's default options, which is equivalent to asking for 1 row and 1 column (i.e. 1 subplot axes).

````python
fig, axes = plt.subplots(2, 1)

axes[0].plot(x, y, lw=3., c='r')
axes[1].plot(x, z, 'b--')

axes[1].set_xlabel("x", fontsize=16)
axes[0].set_ylabel("sin(x)", fontsize=16)
axes[1].set_ylabel(r"sin(x+$\pi$/2)", fontsize=16)
plt.show()
````

Alternatively, we can give a name to each Axes instance, for simpler code later

````python
fig, (ax1, ax2) = plt.subplots(2, 1)

ax1.plot(x, y, lw=3., c='r')
ax2.plot(x, z, 'b--')

ax2.set_xlabel("x", fontsize=16)
ax1.set_ylabel("sin(x)", fontsize=16)
ax2.set_ylabel(r"sin(x+$\pi$/2)", fontsize=16)
plt.show()
````

Or grab all the axes in one object, which we can index later to choose an Axes instance

````python
fig, axes = plt.subplots(2, 2, figsize=(6, 5))

axes[0, 0].plot(x, y, 'r')
axes[0, 1].plot(x, y, 'b')
axes[1, 0].plot(x, y, 'g')
axes[1, 1].plot(x, y, 'k')

plt.show()
````

For more complex subplot layouts, there are several ways to achieve the same thing. One option is to use `plt.subplot2grid`. It maps the figure to a grid, and you can define subplots spanning across several cells of the grid. It takes 4 arguments:
```python
subplot2grid(shape,
             location,
             rowspan = 1,
             colspan = 1)
```
`shape` is the shape of the grid (number of cells), it needs to be the same for each call of the function within one figure. `location` is the location of the top-left corner of the subplot. `rowspan` and `colspan` define respectively the length and heigth of the plot.

````python
fig = plt.figure(figsize=(5, 5))

ax1 = plt.subplot2grid((3, 3), (0, 0))
ax2 = plt.subplot2grid((3, 3), (0, 1), colspan=2)
ax3 = plt.subplot2grid((3, 3), (1, 0), colspan=2, rowspan=2)
ax4 = plt.subplot2grid((3, 3), (1, 2), rowspan=2)
# what happens if you uncomment the following line? 
# ax5 = plt.subplot2grid((3, 3), (1, 1), colspan=1, rowspan=1)

ax1.plot(x, y, 'r')
ax2.plot(x, y, 'b')
ax3.plot(x, y, 'g') 
ax4.plot(x, y, 'k')

plt.show()
````

An alternative is to use a `GridSpec`, which is another way to define a grid layout for subplots within a figure, and lets you control the space around and between subplots and the size ratios of different row and columns.

```
GridSpec(nrows, ncols, figure=None, 
         left=None, bottom=None, right=None, top=None, 
         wspace=None, hspace=None,
         width_ratios=None, height_ratios=None
```

For example:

````python
fig = plt.figure(figsize=(5, 5))

gs = fig.add_gridspec(nrows=2, ncols=3, height_ratios=(1, 2))

ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1:])
ax3 = fig.add_subplot(gs[1, :2])
ax4 = fig.add_subplot(gs[1, 2])

ax1.plot(x, y, 'r')
ax2.plot(x, y, 'b')
ax3.plot(x, y, 'g') 
ax4.plot(x, y, 'k')

plt.show()
````

Finally, `fig.subplot_mosaic` allows you to use nested lists or ASCII art to define the Axes grid positions, for example:

````python
fig = plt.figure(figsize=(5, 5))
ax_dict = fig.subplot_mosaic(
    """
    ABB
    CCD
    CCD
    """
)

ax_dict["A"].plot(x, y, 'r')
ax_dict["B"].plot(x, y, 'b')
ax_dict["C"].plot(x, y, 'g') 
ax_dict["D"].plot(x, y, 'k')
````

### 2D

There are several ways to represent multidimensional data. One can choose to represent the data contained in a 2D domain (a matrix) with colors (in a 2D plot), or with surfaces (in a 3D plot). In the next few slides, we will see several ways to plot 2D data.

````python
# First let's make some data to plot

# Try playing around with these numbers and see what changes
N = 10
x = np.linspace(0,2*np.pi,N)
y = np.sin(x); z = np.cos(x)
#Create 2D field from outer product of previous 1D functions
noise = np.random.random(N**2)
u = np.outer(y, z) + noise.reshape(N, N) / 3
````

````python
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(7, 5), gridspec_kw={"hspace":0.4})

p = ax1.contourf(u, levels=np.linspace(-1, 1.5, 11), cmap="RdYlBu_r")
q = ax2.contour(u, levels=[-1, -0.5, 0, 0.5, 1, 1.5], cmap="RdYlBu_r")
r = ax3.pcolormesh(u, cmap="RdYlBu_r", vmin=-1, vmax=1.5)
s = ax4.imshow(u, cmap="RdYlBu_r", vmin=-1, vmax=1.5)

# Add some titles and colour bars
for ax, mappable, plottype in zip(fig.axes, [p, q, r, s],
                                  ["contourf", "contour", "pcolormesh", "imshow"]):
    fig.colorbar(mappable, ax=ax)
    ax.set_title(plottype)
    ax.set_aspect("equal")

plt.show()
````

### `pcolormesh` or `imshow`

Both these functions allow you to plot a 2D array, but they have some differences. 

- `imshow` is good for plotting pseudocolour images, or for when all of the "cells" in your data have equal size.
- `pcolormesh` allows you to use non-regular rectangular meshes

````python
# if you get a warning about too many open figures, try this:
plt.close("all")
````

````python
fig, ax = plt.subplots()

# Some non-regular coordinates
x = np.logspace(1, 3, 11)
y = np.linspace(0, 1000, 11)
X, Y = np.meshgrid(x, y)

# a fake data array
Z = X[:-1, :-1] + Y[:-1, :-1]

# Note that we define the nodal coordinates, and the cell-centred value of Z
ax.pcolormesh(X, Y, Z)

ax.set_aspect("equal")

ax.set_title("pcolormesh works with non-regular grids")

plt.show()
````

### Animations

Animations (*i.e.* series of images) can be created with matplotlib. However, exercise some restraint as they can require significant resources.

````python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation

fig, ax = plt.subplots()

def get_field(a, N = 100):
    """
    argument "a" will be updated each frame of the animation, from the "frames" iterable
    """
    x = a * np.linspace(0, 2 * np.pi, N)
    y = np.sin(x); z = np.cos(x)
    out = np.outer(y, z)
    ax.contourf(out, 40, cmap="RdYlBu_r")
    ax.set_title(f"a = {a}")
     
ani = matplotlib.animation.FuncAnimation(fig, get_field, frames=range(0, 10),
                                         interval=100)
````

````python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation

try:
    ani.pause()
except:
    pass
plt.close("all")

fig, ax = plt.subplots()

x = y = np.arange(10)
X, Y = np.meshgrid(x, y)

def get_field(a, N = 100):
    """
    argument "a" will be updated each frame of the animation, from the "frames" iterable
    """
    ax.cla()
    ax.quiver(X, Y, X+a, Y-a) 
        
ani = matplotlib.animation.FuncAnimation(fig, get_field, frames=range(0, 10),
                                         interval=100)
plt.show()
````

````python
import random

try:
    ani.pause()
except:
    pass
plt.close("all")

ysample = random.sample(range(-50, 50), 100)

xdata = []
ydata = []

fig, ax = plt.subplots()

ax.set_xlim(0, 100)
ax.set_ylim(-50, +50)
line, = ax.plot(xdata, ydata, 'r-')

def update(i):
    xdata.append(i)
    ydata.append(ysample[i])
    line.set_xdata(xdata)
    line.set_ydata(ydata)

ani = matplotlib.animation.FuncAnimation(fig, update, frames=100,
                                         interval=100, repeat=False)
````

Movies can be generated by saving images then calling an external program like `ffmpeg`, or using the `writers` function from the `animation` submodule, which also uses `ffmpeg` under the hood.

````python
# Close everything now
try:
    ani.pause()
except:
    pass
plt.close("all")
````

### Widgets

Matplotlib provides widgets, that allow you to create dynamic plots with sliders, buttons and other ways to interact with the data. The matplotlib website has [some good examples](https://matplotlib.org/stable/gallery/widgets/index.html) of these, but here's one to get you started.

````python
import matplotlib.widgets as mw

fig, ax = plt.subplots()

# Setup initial plot of sine function
x = np.linspace(0, 2*np.pi, 200)
l, = ax.plot(x, np.sin(x))
ax.set_title(f"sin(1.000x)")

# Adjust figure to make room for slider
fig.subplots_adjust(bottom=0.15)
axslide = fig.add_axes([0.15, 0.05, 0.75, 0.03]) 
s = mw.Slider(axslide, 'A value', 0., 5., valinit=1)

# Define function
def update(A):
    l.set_ydata(np.sin(A * x))
    ax.set_title(f"sin({A:.3f}x)")
    fig.canvas.draw_idle()

    
# Bind update function to change in slider
s.on_changed(update)

plt.show()
````

## Sympy

The `sympy` package provides symbolic algebra routines. That's to say, it gives the ability to process equations in terms of letters & symbols.

````python
import sympy

x, y, z = sympy.symbols('x y z')

m = 2 * x + y + z

print(4 * m)
````

Sympy symbols can be used most places you'd use a number, but may not return the kind of output you'd expect

````python
test = x > 3
print(test) # evaluation is delayed
print(test.subs({x:2})) # substitute a value for x
print(test.subs({x:4})) # substitute a value for x
print(test.subs({x:y})) # Can also substitute symbols
````

We can do many alebraic operations and useful conversions

````python
res = (x + 3) * (x - 4)
print('res:', res)
print('expanded:', res.expand())

res -= 3 * x
print('simplified:', res.simplify())
print('as latex:', sympy.latex(res.simplify()))
````

We can also do basic calculus

````python
f = 3 * x**2 + x

print('f:', f)
print('df_dx:', f.diff(x))
print('int_f:', f.integrate(x))          # indefinite integral
print('int_f:', f.integrate((x, 1, 3)))  # definite integral
````

We can even solve (moderately simple) equations exactly!

````python
a = sympy.Symbol('a')

g = x**2 + a * x + 5

# solve g == 0 for x
print( sympy.solve(sympy.Eq(g, 0), x))

# solve g == 0 for a
print( sympy.solve(sympy.Eq(g, 0), a))
````

Surprising number of uses when you need a weird formula quickly and have a quick way to code it up. Equivalent of a calculator for algebra & calculus. We can also build a matrix and use to calculate things like the determinant:

````python
from pprint import pprint 

mat = []

for i in range(4):
    row = []
    for j in range(4):
        row.append(sympy.Symbol(f'a_{i+1}{j+1}'))
    mat.append(row)

print(mat)

M = sympy.Matrix(mat)
````

````python
pprint(M)
print(f"Determinant: {M.det()}")
````

Unlike `numpy`, `sympy` is not vital for numerical programming in Python, but sometimes it's still very useful.
