## Coding nightmares

### 1. Mars Climate Orbiter



Launched: December 11th 1998
Disintegrated: September 23rd 1999,

During orbital insertion as the planet reached Mars, the trajectory was too steep and too fast.

The orbiter passed through the Martian upper atmosphere and disintegrated.

A [report](ftp://ftp.hq.nasa.gov/pub/pao/reports/1999/MCO_report.pdf) summarised the root of the failure.

#### Cause of failure

Code written by Lockheed Martin gave impulses in pound-force seconds (aagainst specifications) to code, written by NASA expecting figures in Newton seconds.

Roughly 4.45x smaller

The interaction had not been tested.

### 2. The Therac-25



A Canadian X-ray machine controlled by a computer issued doses 100s of times what was intended, killing several people.

The machine had two modes of operation:
- photon therapy (high dose, needing metal shield)
- electron-beam therapy (low dose, without shield)

By switching between modes, software let operators select the high dose mode without the required extra equipment.

#### Causes of the software failure:

- Blind reuse of code from old machines.
- No independent review of software
- No risk assessment
- Use of error codes meaningless to operators
- Lack of testing, particularly of independent components.

### 3. Heartbleed



A bug in a cryptographic library used for internet security allowed multiple attacks on web connected servers.

In 2011 a Ph.D. student submitted new code to the Open Source OpenSSL crypography library. The code was bugged, and improper input validation allowed malicious users unauthorised access to other data in memory.

It's possible that various entities, including the US National Security Agency may have known about the flaw for several months before public reports surfaced and the bug was patched.

#### Causes of the software failure:

- Little risk assessment.
- Excessive use of nonstandard code.
- Lack of robustness testing.
- Overinterest in performace gains above security.

### 4. Excel spreadsheet bugs and Errors

<img src="images/excel.png" alt="Excel logo" width="200"/>

Over the years, various groups of scientists & researchers using Excel (and similar software) have published papers containing errors caused by mistakes in their spreadsheets.

This is both embarrassing and may have led to wrong conclusions.

Notable examples:

- In 2002, a paper in *Nature* reported up to 40% of all papers using gene expression data contained errors, many of which could be traced to Excel mishandling gene names (e.g., converting "SEPT2" to "2-Sep") thanks to autidate.
- In 2010 Harvard economists published a paper 'Growth in a Time of Debt' that contained a critical error in an Excel spreadsheet. The error led to incorrect conclusions about the relationship between national debt and economic growth, which influenced policy debates worldwide. The error left confounding countries (like Australia & Belgium) with high debt but relatively strong growth out of the analysis.

See the European Spreadsheet Risks Interest Group (EuSpRIG) for [more examples](https://www.eusprig.org/horror-stories).

#### Causes and Best Practices

- Problems almost all caused by human errors & lack of validation/verification processes when using spreadsheet software for complex data analysis.

- Mostly identified by independent research groups reviewing published work. Good example of why thorough code review is important.

- Using packages (e.g., R, Python's pandas) can help reduce the risk of such errors:
  - often include built-in functions for data validation and analysis,
  - separate data from code, while keeping both visible.

- If you must use spreadsheets, follow [best practices](https://www.eusprig.org/best-practices.htm):
  - Keep it simple.
  - Document your work thoroughly.
  - [Test your work](https://arxiv.org/abs/0712.0109) with sample data.
