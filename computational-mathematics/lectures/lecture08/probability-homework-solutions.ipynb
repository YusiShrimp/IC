### Motivation

Let's begin by understanding why we even need a dimension reduction technique like PCA. 
Imagine trying to understand or visualise data with many features. 
As the number of dimensions increases, our ability to inspect and understand the data directly becomes incredibly difficult. This is sometimes termed the "Curse of Dimensionality."

To illustrate this, let's make use again of the Iris dataset, which as we've seen has four features. In the grand scheme of things this isn't a large number, but any more than two can be a challenge for us.

We'll try to visualize it in 3D first.

```python
from mpl_toolkits.mplot3d import Axes3D # Required for 3D plots

# Load the Iris dataset
iris = load_iris()
X = iris.data # The features
y = iris.target # The target species
target_names = iris.target_names

# Create a 3D scatter plot
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

colors = ["navy", "turquoise", "darkorange"]

for color, i, target_name in zip(colors, [0, 1, 2], target_names):
    # Plot three of the four features (e.g., sepal length, sepal width, petal length)
    ax.scatter(X[y == i, 0], X[y == i, 1], X[y == i, 2],
               color=color, alpha=0.8, label=target_name)

ax.set_xlabel(iris.feature_names[0])
ax.set_ylabel(iris.feature_names[1])
ax.set_zlabel(iris.feature_names[2])
ax.set_title("Iris Dataset in 3 Dimensions")
ax.legend(loc="best")
plt.tight_layout()
plt.show()
```

Here we can see a 3D plot using three of the Iris features. 
Even with just three dimensions, it's already quite challenging to fully grasp the separation and overlap of the different species. 
Now, imagine if we had all four features, or perhaps tens, hundreds, or even thousands of features - visualising such high-dimensional data is impossible for humans.

This is where PCA comes in. Our goal is to find a way to "squash" or project this high-dimensional data into an appropriately chosen lower-dimensional space (typically 2D or 3D for visualization) while retaining as much of the original data's "information" or variance (spread) as possible. This allows us to see underlying patterns and relationships that would otherwise be hidden.

We will see that simply plotting against any two of the features is not the optimal way to proceed.

### Solution - PCA for the Iris dataset - first approach via the SVD and direct comparison with ScikitLearn

First let's take the SVD route - recall from a previous lecture:

<br>


The [Singular Value Decomposition](https://en.wikipedia.org/wiki/Singular_value_decomposition) generalises the eigen-decomposition concept to non-square matrices.

Consider an arbitrary $m \times n$ real matrix $A$ - very similar to the square case it can be decomposed into a product of three matrices:

$$A = U\Sigma V^{T}$$

where


- $U$ is an $m\times m$ orthogonal matrix - its columns are the eigenvectors of the matrix $AA^T$,


- $V$ is an $n\times n$ orthogonal matrix - its columns are the eigenvectors of the matrix $A^TA$,


- $\Sigma$ is an $m\times n$ diagonal matrix whose diagonal entries, $\sigma_1, \sigma_2,\ldots  $ are what are called the ***singular values*** of $A$.


<br>


<br>

The singular values of $A$ are non-negative, and the convention is to number them/place down the diagonal of $\Sigma$ in order of their magnitude: $\sigma_1\geq\sigma_2\geq\ldots\geq 0$.

<br>

The singular values are the square roots of the eigenvalues of the square matrix $A^TA$.  They are also the square roots of the eigenvalues of the square matrix $AA^T$.  


<br>
<br>

If we expand out this product of three matrices we get

$$A = U\Sigma V^{T} = \sigma_1 u_{:1} v_{:1}^T +  \sigma_2 u_{:2} v_{:2}^T + \ldots $$

```python
# First let's use scikit-learn's PCA functionality to reduce the
# data dimension from 4 features to 2 - then we'll recreate the result
# ourselves without use of scikit-learn

from sklearn.datasets import load_iris
from sklearn import decomposition

# Load the Iris dataset
iris = load_iris()

# Extract the features (data) and the target (species labels)
X = iris.data
y = iris.target

# Set up a figure with two subplots side-by-side
fig = plt.figure(figsize=(12, 5))
ax1 = plt.subplot(121)
ax2 = plt.subplot(122)

# --- Using scikit-learn PCA ---
# Initialize PCA transformer to find the top 2 principal components
pca = decomposition.PCA(n_components=2)
# Compute the principal components based on the data X
pca.fit(X)
# Transform the original data X into the 2-dimensional PCA space
Xp = pca.transform(X)
# Print the shape of the transformed data (should be 150 samples x 2 components)
print(np.shape(Xp))

# Define target names and colors for plotting
target_names = iris.target_names
colors = ["navy", "turquoise", "darkorange"]

# Plot the PCA-transformed data, coloring points by species
for color, i, target_name in zip(colors, [0, 1, 2], target_names):
    # Select data points belonging to the current class 'i'
    # Plot the first principal component (Xp[:, 0]) vs the second (Xp[:, 1])
    ax1.scatter(
        Xp[y == i, 0], Xp[y == i, 1], color=color, alpha=0.8, label=target_name)
ax1.legend(loc="best")
ax1.set_title("PCA(dimensions=2) of the IRIS dataset")
# Add appropriate axis labels for PCA plot
ax1.set_xlabel("Principal Component 1") # Added label
ax1.set_ylabel("Principal Component 2") # Added label

# --- Plotting original first two features for comparison ---
# Plot the original data using only the first two features for comparison
for color, i, target_name in zip(colors, [0, 1, 2], target_names):
    # Select data points belonging to the current class 'i'
    # Plot the first feature (X[:, 0]) vs the second feature (X[:, 1])
    ax2.scatter(
        X[y == i, 0], X[y == i, 1], color=color, alpha=0.8, label=target_name)
ax2.legend(loc="best")
ax2.set_title("Compare with just plotting first two features")
# Add axis labels using feature names
ax2.set_xlabel(iris['feature_names'][0]) # Added label using feature name
ax2.set_ylabel(iris['feature_names'][1]) # Added label using feature name
```

Comparing the two plots, we can see the effect of PCA. The right plot simply uses the first two original features (sepal length and width). The left plot uses the two principal components derived from all four original features. Notice how the PCA plot (left) achieves better separation between the three classes (especially between turquoise and orange) compared to just using the first two original features (right). This is because PCA finds the new axes (principal components) that capture the maximum variance in the data, often revealing underlying structures and making clusters more distinct. This improved separation in the reduced space is a key benefit of dimension reduction techniques like PCA for visualization and subsequent classification tasks.

```python
# let's try to demystify what's actually gone on "under the hood"
# to achieve this result - let's do this by recreating manually from our
# knowledge of the SVD from linear algebra, and see what we need to do to exactly recreate
# what scikit-learn keeps largely hidden from us

# Reload original data just in case
X = iris.data
y = iris.target

# Based on L3 homework:
import scipy.linalg as sl
from pprint import pprint

# --- Manual PCA via SVD ---
# Step 1: Center the data by subtracting the mean of each feature (column)
# Calculate the mean of each column (axis=0), keepdims ensures result is broadcastable
X_mean = X.mean(axis=0, keepdims=True)
# Subtract the mean from the original data
Xc = X - X_mean

# Step 2: Perform Singular Value Decomposition (SVD) on the centered data
# U: Left singular vectors (shape m x m)
# S: Singular values (vector of length min(m, n))
# VT: Transposed right singular vectors (shape n x n)
UU, SS, VVT = sl.svd(Xc)

# Reconstruct the diagonal Sigma matrix from the singular values vector SS
# Initialize Sigma with zeros, matching the shape of Xc (m x n)
SSS = np.zeros_like(Xc, dtype='float')
# Get the minimum dimension for placing singular values
min_dim = min(Xc.shape[0], Xc.shape[1])
# Place singular values onto the main diagonal of SSS
SSS[:min_dim, :min_dim] = np.diag(SS)

# Check that the SVD reconstruction Xc = U * Sigma * V^T holds true
print('SVD success: ',np.allclose(Xc, UU @ SSS @ VVT))

# Print the singular values (related to variance explained by each component)
print('singular vals = ', SS)

# Step 3: Project the data onto the first two principal components
# The principal components are the first 'n_components' columns of V (or rows of VT)
# Select the first 2 rows of VT and transpose to get the first 2 principal directions (shape 4 x 2)
principal_directions = VVT[:2, :].T
# Print shapes to verify dimensions for matrix multiplication
print(np.shape(Xc))                  # Shape (150 x 4)
print(np.shape(principal_directions)) # Shape (4 x 2)

# Project the centered data (Xc) onto the principal directions
# through the following multiplication
Xp = Xc @ principal_directions  # Resulting shape should be (150 x 2)
# Print the shape of the projected data to confirm
print(np.shape(Xp))

# --- Plot the manually PCA-transformed data ---
# Set up the plot
fig = plt.figure(figsize=(6, 5))
ax1 = plt.subplot(111)

# Define target names and colors again
target_names = iris.target_names
colors = ["navy", "turquoise", "darkorange"]

# Plot the projected data, coloring points by species
for color, i, target_name in zip(colors, [0, 1, 2], target_names):
    # Select projected data points belonging to the current class 'i'
    # Plot the first component (Xp[:, 0]) vs the second component (Xp[:, 1])
    ax1.scatter(
        Xp[y == i, 0], Xp[y == i, 1], color=color, alpha=0.8, label=target_name)
ax1.legend(loc="best")
ax1.set_title("Our PCA via SVD")
```

## Homework - Die-rolling experiment

Investigate how the probabilities computed for the die-rolling experiment equalise as you increase the number of times you roll the die.

### Solution - Die-rolling experiment


```python
# Function to simulate n die rolls
def dice_experiment(n):
    # Set the seed if you want to make the random experiment reproducible.
    # If you exclude this you should get a different result each time you run.
    #np.random.seed(1)
    # Rolls are integers from 1 (inclusive) up to 7 (exclusive) ...
    # return n rolls from the S = {1, 2, 3, 4, 5, 6} sample space
    return np.random.randint(low=1, high=7, size=n)

# Function to calculate empirical probabilities from outcomes
def calculate_probabilities(outcomes):
    # Find unique outcomes and how many times each occurred
    values, counts = np.unique(outcomes, return_counts=True)
    total_outcomes = len(outcomes)
    probabilities = counts / total_outcomes
    # Ensure all outcomes 1-6 are represented, even if they had 0 count
    prob_dict = dict(zip(values, probabilities))
    full_values = np.arange(1, 7)
    full_probabilities = np.array([prob_dict.get(i, 0) for i in full_values])
    return full_values, full_probabilities

# Function to plot the PMF using a stem plot
def plot_pmf(values, probabilities, title="Probability Mass Function"):
    plt.figure()
    # Use stem plot for discrete distributions
    plt.stem(values, probabilities, basefmt=" ")
    plt.xlabel("Outcome (x)")
    plt.ylabel("Probability P(x)")
    plt.title(title)
    # Set y-limit to make comparisons easier, e.g., slightly above 1/6
    plt.ylim(0, 0.3)
    # Ensure x-axis ticks show all integer outcomes
    plt.xticks(np.arange(1, 7))
    plt.grid(True, axis='y')
    plt.show()
    
```

```python
# Simulate 1 million rolls
n_rolls_large = 1_000_000
outcomes_large = dice_experiment(n_rolls_large)

# Calculate and plot for the large number of rolls
values_large, probs_large = calculate_probabilities(outcomes_large)
print(f"\nEmpirical probabilities from {n_rolls_large} rolls:")
for v, p in zip(values_large, probs_large):
    print(f" P({v}) = {p:.6f} (Theoretical = {1/6:.6f})")

plot_pmf(values_large, probs_large, title=f"Empirical PMF from {n_rolls_large} Die Rolls")

# Check that the probabilities sum to 1
print(f"\nSum of probabilities: {np.sum(probs_large):.4f}")


# we could gauge how equal these are by finding the different between the max 
# probability and the smallest
print(np.max(probs_large) - np.min(probs_large))
```

## Homework - The Gaussian integral

A useful definite integral to know is the area under the function $e^{-x^2}$ over the whole real line:

$$\int_{-\infty}^{\infty} e^{-x^2}\, dx = \sqrt{\pi}$$

This result is known as the [Gaussian integral](https://en.wikipedia.org/wiki/Gaussian_integral).

Read the description at the above link for how this integral can be (relatively) easily established if you know about polar coordinates and know the "trick".

<br>

Once we have this key result we can establish that 

$$\int_{-\infty}^{\infty} e^{-\frac{1}{2} \left( \frac{x-\mu}{\sigma} \right)^2 }\, dx = \sigma\sqrt{2\pi}$$

and thus by dividing through by the quantity on the RHS we have a function

$$f(x) = \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{1}{2} \left( \frac{x-\mu}{\sigma} \right)^2 }$$

(note that this is the [**Gaussian function**](https://en.wikipedia.org/wiki/Gaussian_function)) which has the property that its integral over the entire real line is unity.

<br>

Can you see how to establish that 

$$\int_{-\infty}^{\infty} e^{-x^2}\, dx = \sqrt{\pi}  
\implies
\int_{-\infty}^{\infty} e^{-\frac{1}{2} \left( \frac{x-\mu}{\sigma} \right)^2 }\, dx = \sigma\sqrt{2\pi}
$$

through two changes of variables?

### Solution - The Gaussian integral

We do this by change of variables. Starting from

$$\int_{-\infty}^{\infty} e^{-x^2}\, dx = \sqrt{\pi} $$

introduce the new variable

$$y:= \sqrt{2} x \implies x = \frac{y}{\sqrt{2}}, 
\quad \text{and}\quad \frac{dy}{dx}={\sqrt{2}}\implies dx = \frac{1}{\sqrt{2}}dy
$$

we also need to remember to note that the limits of the integral $x\rightarrow \pm\infty$ transform to $y\rightarrow \pm\infty$.  We can then transform the variables in the integral:

$$\sqrt{\pi} = \int_{-\infty}^{\infty} e^{-x^2}\, dx = 
\int_{-\infty}^{\infty} e^{-\frac{1}{2} y^2}\, \frac{1}{\sqrt{2}}\, dy
=\frac{1}{\sqrt{2}} \int_{-\infty}^{\infty} e^{-\frac{1}{2} y^2}\, dy
$$ 

and hence

$$\int_{-\infty}^{\infty} e^{-\frac{1}{2} y^2} \, dy = \sqrt{2} \sqrt{\pi}
=\sqrt{2\pi}$$ 

Note that $y$ here is a [*dummy variable*](https://mathworld.wolfram.com/DummyVariable.html) 
and so we could replace it here with $x$ (which in the original integral was also a dummy variable of course) if we wanted, but we won't do this yet as we're not finished.

Let's make the further change of variables

$$z:= \sigma\, y + \mu \implies y = \frac{z - \mu}{\sigma}, 
\quad \text{and}\quad \frac{dz}{dy}=\sigma \implies dy = \frac{1}{\sigma}dz
$$

again we need to note that  $y\rightarrow \pm\infty$ transforms to $z\rightarrow \pm\infty$ (as we can assume that $\sigma>0$).

We therefore have that 

$$\sqrt{2\pi} = \int_{-\infty}^{\infty} e^{-\frac{1}{2} y^2} \, dy 
=\int_{-\infty}^{\infty} e^{-\frac{1}{2} \left(\frac{z - \mu}{\sigma}\right)^2} \,\frac{1}{\sigma}\, dz 
$$

and therefore 

$$\int_{-\infty}^{\infty} e^{-\frac{1}{2} \left(\frac{z - \mu}{\sigma}\right)^2} \, dz 
= \sigma\sqrt{2\pi}
$$

We can replace our dummy variable ($z$ here) with a different symbol (let's choose $x$), and we have that 

$$\frac{1}{\sigma\sqrt{2\pi}}\int_{-\infty}^{\infty} e^{-\frac{1}{2} \left(\frac{x - \mu}{\sigma}\right)^2} \, dx 
= 1
$$

i.e. the function

$$f(x) := \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{1}{2} \left( \frac{x-\mu}{\sigma} \right)^2 }$$

has the property that 

$$\int_{-\infty}^{\infty} f(x) \, dx = 1 $$

## Homework -  Maximum likelihood estimation  [$\star\star$]

Investigate what happens if we don't use all 100 chips in the MLE example from class.

Generalise the example so you seek to minimise the negative log likelihood instead of maximising the likelihood.

### Solution -  Maximum likelihood estimation

```python
import random
import math

random.seed(42)


def draw(frac_red, k=20):
    # draw k chips from the bag comprising frac_red*100 red chips
    # and (1-frac_red)*100 green chips
    return random.sample(['red']*int(frac_red*100) + ['green']*int((1-frac_red)*100), k=k)


def calc_prob_single(outcome, frac_red):
    # calculate the probability of "outcome" occurring given
    # fraction of red chips 
    assert(outcome in ('red', 'green'))
    if outcome == "red":
        return frac_red
    else:
        return 1 - frac_red


def update_frac(frac_red, total, result):
    # update the remaining chips in order to update the
    # fraction red value; total is the number of chips in the bag *before* this draw
    # (round, not int: frac_red * total is the count up to round-off, e.g. 57.99999...)
    nums = [round(frac_red * total), round((1-frac_red) * total)]

    if result == "red":
        nums[0]-= 1
    else:
        nums[1]-= 1
    return float(nums[0]/sum(nums))


def calc_prob(results, frac_red, total):
    prob = 1
    for result in results:
        p = calc_prob_single(result, frac_red)
        prob *= p
        frac_red = update_frac(frac_red, total, result)
        total -= 1  # one chip fewer in the bag
        
    return prob


TRUE_PROPORTION = 0.6

# Here we draw 100 chips from the bag, using the true proportion.
results = draw(TRUE_PROPORTION, k=100)

# From this point forth, we will only use the drawn chips in an attempt
# to estimate the true proportion of chips. 

# Possible values of the proportion: 0.0, 0.1, 0.2, 0.3, ...., 0.9, 1.0
allks = [x/10 for x in range(0, 11)]

# Calculate the probability of observing this data (i.e. likelihood), 
# for each value of the proportion
# use n_chips
n_chips = 20
probs = [calc_prob(results[0:n_chips], x, 100) for x in allks]

print(probs)

print('Likelihood maximised at:', allks[np.argmax(probs)])

plt.clf()
plt.plot(allks, probs, 'k.-')
plt.show()
```

```python
from math import log

def calc_nll(results, frac_red, total):
    nll = 0
    for result in results:
        p = calc_prob_single(result, frac_red)
        # avoid taking log of zero
        nll += -log(max(p, 1.e-10))
        frac_red = update_frac(frac_red, total, result)
        total -= 1  # one chip fewer in the bag
        
    return nll
```

```python
fig = plt.figure(figsize=(10, 6))
ax1 = plt.subplot(111)

ax1.set_title('Logistic function', fontsize=16)
ax1.set_xlabel('$x$', fontsize=16)
ax1.set_ylabel(r'$\sigma(x)$', fontsize=16)

def logistic_fun(x):
    return 1 / (1 + np.exp(-x))

# define our x for plotting purposes
x = np.linspace(-10., 10., 1000)

# plot our function
ax1.plot(x, logistic_fun(x), 'k', lw=2, label='Logistic function')
```

A derivation of the derivative can be seen here <https://en.wikipedia.org/wiki/Logistic_function#Derivative> and requires use of the [quotient rule](https://en.wikipedia.org/wiki/Quotient_rule) of differentiation after noting that 

$$
\sigma(z) := \frac{1}{1+e^{-z}} \equiv \frac{e^{z}}{e^{z}+1}
$$

Then 

$$
\frac{d \sigma(z)}{dz} := \frac{e^{z}}{(1+e^{z})^2}
$$

```python
fig = plt.figure(figsize=(10, 6))
ax1 = plt.subplot(111)

ax1.set_title('d(Logistic function)/dx', fontsize=16)
ax1.set_xlabel('$x$', fontsize=16)
ax1.set_ylabel(r'$d\sigma(x)/dx$', fontsize=16)

def dlogistic_fun(x):
    return np.exp(x) / (1 + np.exp(x))**2

# define our x for plotting purposes
x = np.linspace(-10., 10., 1000)

# plot our function
ax1.plot(x, dlogistic_fun(x), 'k', lw=2, label='Logistic function derivative')
```

```python
fig = plt.figure(figsize=(10, 6))
ax1 = plt.subplot(111)

ax1.set_title('Logistic function and its derivative', fontsize=16)
ax1.set_xlabel('$x$', fontsize=16)

# define our x for plotting purposes
x = np.linspace(-10., 10., 1000)

# plot our functions
ax1.plot(x, logistic_fun(x), 'k', lw=2, label='Logistic function')
ax1.plot(x, dlogistic_fun(x), 'r', lw=2, label='Logistic function derivative')
ax1.legend(loc='best', fontsize=10)
```

To shift and tighten the jump we need to make two changes to the formula:

```python
fig = plt.figure(figsize=(10, 6))
ax1 = plt.subplot(111)

ax1.set_title('Logistic function', fontsize=16)
ax1.set_xlabel('$x$', fontsize=16)
ax1.set_ylabel(r'$\sigma(x)$', fontsize=16)

def logistic_fun_shift_tight(x):
    return 1 / (1 + np.exp(-10*(x-5)))

# define our x for plotting purposes
x = np.linspace(-10., 10., 1000)

# plot our function
ax1.plot(x, logistic_fun_shift_tight(x), 'k', lw=2, label='Logistic function')
```

## Monte Carlo methods [\*]

In general, Monte Carlo methods aim to use stochastic/probabilistic methods to estimate something that is deterministic (i.e. not stochastic). 

There is no general agreement on how Monte Carlo methods should be defined.  
Here is an attempt to define *Monte Carlo simulation* through an example of coin tossing (from <https://en.wikipedia.org/wiki/Monte_Carlo_method#Definitions>):


- **Simulation**: Drawing *one* [pseudo-random](https://en.wikipedia.org/wiki/Pseudorandomness) uniform variable $x$ from the interval $[0, 1]$ can be used to simulate the tossing of a coin: If $x \le 0.5$, we designate the outcome as "heads", but if $x > 0.5$, we designate it as tails. This is a simulation, but not a Monte-Carlo simulation.


- **Monte Carlo method**: Pouring out a box of coins on a table, and then computing the ratio of coins that land "heads" vs "tails" is a Monte Carlo method to estimate the probabilities of "heads" and "tails" in a coin toss, but it is not a simulation. 


- **Monte Carlo simulation**: Drawing a *large number* of pseudo-random uniform variables from the interval $[0, 1]$ at one time, or once at many different times, and assigning values less than or equal to 0.5 as "heads" and greater than 0.5 as "tails", is a *Monte Carlo simulation* of the behaviour of repeatedly tossing a coin.

### Example - Estimating the value of $\pi$ using a Monte Carlo method

One method to estimate the value of $\pi = 3.141592\ldots$ is by using a Monte Carlo method. 

This method consists of drawing a square with an inner circle. We then generate a large number of random points within the square and count how many fall in the enclosed circle:

<img src="images/pi-monte-carlo.png" />

<br>

By knowing the area of the circle $A_\mathrm{c} = \pi r^2$ and the area of the square $A_\mathrm{s} = (2r)^2 = 4r^2$, we can compute the probability of a random point falling both in the square and in the circle as the ratio of the size of the circle to that of the square:

$$P = \frac{A_\mathrm{c}}{A_\mathrm{s}} = \frac{\pi r^2}{4r^2} = \frac{\pi}{4}$$

Hence, we generate a large number of random points inside the square and count how many fall inside the circle. 

$$\pi = 4P = 4 \cdot \frac{\text{points inside the circle}}{\text{total points}}$$

</div>

```python
import random

n_total = 100_000  # total number of points
x = np.random.uniform(low=-1, high=1, size=n_total)
y = np.random.uniform(low=-1, high=1, size=n_total)

z = np.sqrt((x**2) + (y**2))  # distance from the origin (0, 0)

n_circle = sum(z<=1)  # number of points falling in the circle
pi_estimate = 4 * n_circle/n_total

print(f'The estimate of pi is {pi_estimate}.')
```

Let's look at the simulated points:

```python
# Label points
in_out = np.where(z<=1, 'in', 'out')

# Plot points
plt.figure(figsize=(10, 10))
sns.scatterplot(x=x, y=y, hue=in_out, legend=False)

# Axes labels
plt.xlabel('x')
plt.ylabel('y')

plt.show()
```

#### Solution - Monte-Carlo simulation

This time we consider a sphere inside a box.

We generate a large number of random points within the box and count how many fall inside the sphere.

Let's consider the unit box with one corner at the origin $(0,0,0)$ and the other corners at
$(1,0,0)$, $(0,1,0)$, ..., $(1,1,1)$.

The sphere that sits inside this box is centred at $(1/2, 1/2, 1/2)$ and has radius 1/2.

The volume of the box is 1, the volume of the sphere is $(4/3)\pi r^3 = (4/3)\pi (1/8) = \pi/6$,
and hence the ratio of the sphere to the box is also $\pi/6$.

We therefore expect the probability $P$ of each point that lies inside the box to also lie in the sphere to be $\pi/6$ and so

$$\pi = 6P = 6 \cdot \frac{\text{points inside the sphere}}{\text{total points}}$$

```python
import random

n_total = 1_000_000  # total number of points, selected to lie in box
x = np.random.uniform(low=0, high=1, size=n_total)
y = np.random.uniform(low=0, high=1, size=n_total)
z = np.random.uniform(low=0, high=1, size=n_total)

# square of distance from the centre of the sphere, (1/2, 1/2, 1/2)
r2 = ((x-0.5)**2) + ((y-0.5)**2) + ((z-0.5)**2) 

n_circle = sum(r2 <= (0.5**2))  # number of points falling in the sphere
pi_estimate = 6 * n_circle/n_total

print(f'The estimate of pi is {pi_estimate}.')
```

## An Introduction to Gaussian Process Regression [\*]

We've seen linear regression, both as fitting a line by minimising squared errors and probabilistically by assuming errors follow a normal distribution. This works well for linear trends but struggles with complex, non-linear patterns. Also, standard linear regression gives a single prediction value, but often we want to know how *confident* we are in that prediction.

**Gaussian Process Regression (GPR)** offers a powerful alternative that directly addresses these points. It's a non-parametric, Bayesian approach to regression.

**Motivation:** Why consider GPR?

1.  **Non-linearity:** GPR can naturally model complex, non-linear relationships in data without us having to specify the exact form of the function beforehand (unlike polynomial regression where we choose the degree).
2.  **Uncertainty Quantification:** A key strength of GPR is that it doesn't just give a single prediction ('point estimate'); it provides a full probability distribution for its prediction. This means we get both the most likely value (the mean) *and* a measure of uncertainty (the variance or standard deviation) around that prediction. This is crucial for understanding the reliability of the model, especially in areas where data is sparse.
3.  **Bayesian Approach:** It fits naturally into the Bayesian framework, allowing prior knowledge about the function's behaviour (like smoothness) to be incorporated.

---

**Core Idea (Simple Terms): A Distribution Over Functions**

Instead of assuming a specific *form* for our function (like $y = \beta_0 + \beta_1 x$) and trying to find the best *parameters* ($\beta_0, \beta_1$), GPR takes a different approach. It defines a **probability distribution over *all possible functions*** that might fit the data.

* Think of it like this: Before seeing any data, GPR assumes that any smooth function is possible.
* When we provide data points, GPR updates this distribution. GPR uses Bayes' theorem to update this "prior" distribution over functions to a "posterior" distribution that incorporates the observed data. Functions that pass close to the data points become more probable, and functions that are far away become less probable.
* Predictions are made by averaging over all the plausible functions, weighted by their probability. The variation among these plausible functions gives us the uncertainty estimate.

**How is this distribution defined?**

A Gaussian Process (GP) is defined by two components:

1.  **Mean Function $m(x)$:** This specifies the average or expected value of the function at any point $x$. It often represents our prior belief about the function's baseline. For simplicity, it's very commonly set to zero, assuming that, before seeing data, the function is equally likely to be above or below zero.
2.  **Covariance Function (Kernel) $k(x, x')$:** This is the heart of the GP. The kernel function defines the **covariance** between the function's values at any two points, $x$ and $x'$. It essentially measures how "similar" or related the function's output $f(x)$ is expected to be to $f(x')$, based on the distance between $x$ and $x'$.
    * If $x$ and $x'$ are close, the kernel typically returns a high covariance, meaning $f(x)$ and $f(x')$ are expected to be similar.
    * If $x$ and $x'$ are far apart, the covariance is low, meaning $f(x)$ and $f(x')$ are less related.
    * The *choice* of kernel function encodes our assumptions about the function we're trying to model (e.g., how smooth it is, whether it's periodic). A common choice is the Radial Basis Function (RBF) kernel (also called the squared exponential or Gaussian kernel), which favours smooth functions.

**Making Predictions:**

Given some training data points $(X_{train}, Y_{train})$, GPR uses the GP prior (defined by the mean and kernel) and the data to calculate the *posterior distribution* at new test points $X_{test}$. Crucially, this posterior distribution is also Gaussian. This means for any test point $x_*$, the prediction is not just a single value $y_*$ but a normal distribution:

$$y_* | X_{train}, Y_{train}, x_* \sim \mathcal{N}(\mu_*, \sigma_*^2)$$

* $\mu_*$ is the predicted mean value at $x_*$. This is usually taken as the GPR prediction.
* $\sigma_*^2$ is the predicted variance at $x_*$. The standard deviation $\sigma_*$ gives us the uncertainty bounds (e.g., $\mu_* \pm 2\sigma_*$ represents an approximate 95% confidence interval). The variance tends to be smaller near the training data points and larger further away.

*(The actual formulas for $\mu_*$ and $\sigma_*^2$ involve matrix operations using the kernel function evaluated at the training and test points, but the key idea is that they can be calculated directly once the kernel and training data are known.)*

</div>



**Code Example using Scikit-learn:**

Let's fit a GP model to some simple non-linear data. We'll use `scikit-learn`'s Gaussian Process Regressor.



```python

from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel as C

# --- Generate some non-linear sample data ---
np.random.seed(1) # for reproducibility
n_samples = 10
X_train = np.linspace(0, 10, n_samples).reshape(-1, 1) # Training points
# A non-linear function + some noise
y_train = np.sin(X_train).ravel() + np.random.normal(0, 0.2, X_train.shape[0])

# Points where we want to make predictions
X_test = np.linspace(0, 10, 100).reshape(-1, 1)

# --- Define the GP Kernel ---
# We'll use a common kernel: a constant factor times an RBF kernel.
# The RBF kernel determines the smoothness/length-scale of the function.
# The ConstantKernel allows the overall magnitude to vary.
# length_scale=1.0 is an initial guess; the GPR will optimize these parameters.
kernel = C(1.0, (1e-3, 1e3)) * RBF(length_scale=1.0, length_scale_bounds=(1e-2, 1e2))

# --- Create and Fit the GPR Model ---
# alpha=0.2**2 adds the assumed noise level (variance) to the diagonal of the kernel matrix during fitting.
gp = GaussianProcessRegressor(kernel=kernel, alpha=0.2**2, n_restarts_optimizer=10)

gp.fit(X_train, y_train)

# --- Make Predictions ---
# predict returns both the mean prediction and the standard deviation
y_pred_mean, y_pred_std = gp.predict(X_test, return_std=True)

# --- Plot the Results ---
plt.figure(figsize=(8, 6))
# Plot observed data
plt.scatter(X_train, y_train, c='red', marker='x', s=50, label='Observations')
# Plot GPR mean prediction
plt.plot(X_test, y_pred_mean, 'b-', label='GPR Mean Prediction')
# Plot uncertainty bounds (+/- 1.96 standard deviations for 95% confidence)
plt.fill_between(X_test.ravel(),
                 y_pred_mean - 1.96 * y_pred_std,
                 y_pred_mean + 1.96 * y_pred_std,
                 alpha=0.2, color='blue', label='95% Confidence Interval')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Gaussian Process Regression Example")
plt.legend()
plt.grid(True)
plt.show()

print(f"Optimized Kernel: {gp.kernel_}")
print(f"Log-Marginal-Likelihood: {gp.log_marginal_likelihood(gp.kernel_.theta):.3f}")
```

**Explanation of the Plot:**

* The red 'x' marks are our noisy training data points.
* The solid blue line is the **mean prediction** from the GPR model. Notice how it captures the underlying sine wave shape reasonably well, even though we only gave it 10 noisy data points.
* The shaded blue area represents the **uncertainty**. It's calculated using the standard deviation returned by the GPR (`y_pred_std`). The uncertainty is lowest (the shaded area is narrowest) near the observed data points, and it grows in the gaps between them. Outside the range of the data (beyond $x=10$, say) it would grow quickly towards the prior uncertainty. This intuitive behaviour is a major advantage of GPR.

Gaussian processes are a powerful tool often used in machine learning, optimisation (Bayesian optimisation), and spatial statistics. They provide a flexible way to model functions and explicitly represent the uncertainty in our predictions.
